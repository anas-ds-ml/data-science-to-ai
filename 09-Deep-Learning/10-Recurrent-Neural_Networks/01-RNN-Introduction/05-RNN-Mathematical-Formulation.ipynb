# RNN Mathematical Formulation

A Recurrent Neural Network (RNN) processes sequential data one time step at a time.

At each time step, the RNN takes:

1. The current input `xₜ`
2. The previous hidden state `hₜ₋₁`

and produces:

1. A new hidden state `hₜ`
2. An output `yₜ`

The overall idea is:

```text
Current Input (xₜ)
       +
Previous Hidden State (hₜ₋₁)
       ↓
   RNN Computation
       ↓
New Hidden State (hₜ)
       ↓
   Output Layer
       ↓
Output (yₜ)

# RNN Mathematical Formulation

## 1. Hidden State Equation

The main equation of a basic RNN is:

$$
h_t = \tanh(W_{xh}x_t + W_{hh}h_{t-1} + b_h)
$$

Where:

- $x_t$ = Current input at time step $t$
- $h_{t-1}$ = Previous hidden state
- $h_t$ = New hidden state
- $W_{xh}$ = Input-to-hidden weight matrix
- $W_{hh}$ = Hidden-to-hidden (recurrent) weight matrix
- $b_h$ = Hidden-state bias
- $\tanh$ = Activation function

---

## 2. Output Equation

The output of the RNN is calculated using:

$$
y_t = W_{hy}h_t + b_y
$$

Where:

- $h_t$ = Current hidden state
- $W_{hy}$ = Hidden-to-output weight matrix
- $b_y$ = Output bias
- $y_t$ = Output at time step $t$

---

## 3. Complete RNN Formulation

The complete mathematical flow is:

$$
h_t = \tanh(W_{xh}x_t + W_{hh}h_{t-1} + b_h)
$$

$$
y_t = W_{hy}h_t + b_y
$$

Therefore:

$$
\boxed{x_t + h_{t-1} \rightarrow h_t \rightarrow y_t}
$$

---

## 4. Initial Hidden State

At the beginning of the sequence, there is no previous hidden state, so we commonly initialize:

$$
h_0 = 0
$$

The first hidden state is therefore:

$$
h_1 = \tanh(W_{xh}x_1 + W_{hh}h_0 + b_h)
$$

Then the hidden state is passed forward:

$$
h_1 \rightarrow h_2 \rightarrow h_3 \rightarrow \cdots \rightarrow h_t
$$

````python
import numpy as np

# -----------------------------
# Input and previous hidden state
# -----------------------------
x_t = 2.0
h_previous = 1.0

# -----------------------------
# RNN parameters
# -----------------------------
W_xh = 0.5
W_hh = 0.8
b_h = 0.1

# -----------------------------
# Step 1: Weighted input
# -----------------------------
input_part = W_xh * x_t

# -----------------------------
# Step 2: Weighted previous memory
# -----------------------------
memory_part = W_hh * h_previous

# -----------------------------
# Step 3: Add everything
# -----------------------------
z_t = input_part + memory_part + b_h

# -----------------------------
# Step 4: Apply tanh
# -----------------------------
h_t = np.tanh(z_t)

print("Weighted input       :", input_part)
print("Weighted memory      :", memory_part)
print("Bias                 :", b_h)
print("Before tanh (z_t)    :", z_t)
print("New hidden state     :", h_t)
````

````python
import numpy as np

# Input sequence
sequence = [1.0, 2.0, 3.0, 4.0]

# RNN parameters
W_xh = 0.5
W_hh = 0.8
b_h = 0.1

# Initial hidden state
h = 0.0

print("RNN Sequence Processing\n")

for t, x_t in enumerate(sequence, start=1):

    # RNN mathematical formula
    h = np.tanh(W_xh * x_t + W_hh * h + b_h)

    print(f"Time Step {t}")
    print(f"Input x{t}           = {x_t}")
    print(f"Hidden State h{t}    = {h:.4f}")
    print("-" * 35)
````
