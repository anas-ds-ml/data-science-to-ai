# Artificial Neural Network (ANN) Architecture

An Artificial Neural Network (ANN) is a computational model
made up of interconnected artificial neurons.

These neurons are organized into layers.

A basic ANN contains:

1. Input Layer
2. Hidden Layer(s)
3. Output Layer

Basic architecture:

Input Layer
     ↓
Hidden Layer
     ↓
Output Layer

# 1. Input Layer

The input layer receives the input features from the dataset.

For example, suppose we want to predict house prices.

Features might include:

- Area
- Number of bedrooms
- Number of bathrooms
- Location score

Then the input layer contains one input for each feature.

Example:

x₁ = Area
x₂ = Bedrooms
x₃ = Bathrooms
x₄ = Location Score

Therefore:

Number of input features = Number of input neurons

## Important Point About the Input Layer

The input layer usually does not perform the main mathematical
transformation itself.

It represents the features entering the neural network.

Example:

Dataset:

4 Features
     ↓
Input Layer
     ↓
4 Input Values

If X has shape:

(number_of_samples, 4)

then the network expects:

4 input features

# 2. Hidden Layers

Hidden layers are the layers between the input layer and
the output layer.

Example:

Input Layer
     ↓
Hidden Layer 1
     ↓
Hidden Layer 2
     ↓
Output Layer

A hidden layer contains multiple artificial neurons.

Each neuron can learn different patterns from the input data.

## Why Are Hidden Layers Important?

Hidden layers allow the network to learn complex relationships.

For example:

Input Features
      ↓
Simple Patterns
      ↓
More Complex Patterns
      ↓
High-Level Patterns
      ↓
Prediction

Early layers can learn simpler relationships while deeper
layers can combine those relationships into more complex patterns.

# 3. Output Layer

The output layer produces the final prediction of the neural network.

The number of output neurons depends on the problem.

### Binary Classification

Usually:

1 output neuron

Example:

Spam = 1
Not Spam = 0

Activation:

Sigmoid

### Multiclass Classification

Usually:

One output neuron per class

Example:

Cat
Dog
Horse

Output neurons = 3

Activation:

Softmax

### Regression

Usually:

1 output neuron

Activation:

Linear

# Neuron Calculation Inside a Layer

Every neuron performs a mathematical operation.

The basic equation is:

z = Σ(wᵢxᵢ) + b

Then an activation function is applied:

a = f(z)

Therefore:

Inputs
   ↓
Weights
   ↓
Weighted Sum + Bias
   ↓
Activation Function
   ↓
Neuron Output

# Deep Neural Network

If an ANN contains multiple hidden layers, it is commonly
called a Deep Neural Network (DNN).

Example:

Input
  ↓
Hidden Layer 1
  ↓
Hidden Layer 2
  ↓
Hidden Layer 3
  ↓
Hidden Layer 4
  ↓
Output

The term "deep" refers to having multiple layers of
learnable transformations.

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    
    tf.keras.layers.Dense(16, activation="relu"),
    
    tf.keras.layers.Dense(8, activation="relu"),
    
    tf.keras.layers.Dense(1, activation="sigmoid")
])

model.summary()
```

# Understanding Dense Layer

A Dense layer is a fully connected neural network layer.

Example:

Dense(16, activation="relu")

Means:

16 = Number of neurons

activation="relu"
= Activation function used by the neurons

Every neuron in this Dense layer is connected to every neuron
in the previous layer.

# Parameters in an ANN

The trainable parameters of an ANN are mainly:

1. Weights
2. Biases

For a Dense layer:

Parameters = (Input Features × Neurons) + Neurons

The second term represents the bias for each neuron.

Example:

Input Features = 4
Neurons = 16

Weights:

4 × 16 = 64

Biases:

16

Total:

64 + 16 = 80 parameters
