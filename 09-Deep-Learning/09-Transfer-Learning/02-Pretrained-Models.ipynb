# Pretrained Models in Transfer Learning

A pretrained model is a Deep Learning model that has already been trained on a large dataset.

Instead of starting with random weights, we can load the learned weights of the pretrained model and reuse them for our own task.

In computer vision, many pretrained CNN models have been trained on ImageNet.

## What is ImageNet?

ImageNet is a very large image dataset containing millions of images belonging to thousands of categories.

Examples of categories include:

- Dogs
- Cats
- Cars
- Birds
- Fruits
- Vehicles
- Animals
- Objects

Models such as VGG, ResNet, MobileNet and EfficientNet have commonly been pretrained on ImageNet.

Because these models have seen a huge variety of images, they have learned many useful visual features.

## Why Are Pretrained Weights Useful?

Consider the first layers of a CNN.

They may learn things like:

- Vertical edges
- Horizontal edges
- Curves
- Corners
- Textures

These features are useful for many different image problems.

For example:

A model trained on ImageNet may learn edge and texture features.

We can reuse those features for:

ImageNet
   ↓
Pretrained CNN
   ↓
Learned visual features
   ↓
Our Cat vs Dog dataset

## 1. VGG16

VGG16 is a classic CNN architecture.

It contains many convolutional layers followed by fully connected layers.

Basic idea:

Input Image
    ↓
Convolution
    ↓
Convolution
    ↓
Pooling
    ↓
Convolution
    ↓
Convolution
    ↓
Pooling
    ↓
...
    ↓
Fully Connected Layers
    ↓
Classification

VGG16 is relatively simple to understand, but it has a large number of parameters and is computationally expensive compared with newer architectures.

## 2. VGG19

VGG19 is similar to VGG16 but contains more layers.

VGG16 → 16 weight layers

VGG19 → 19 weight layers

The architecture is based mainly on:

Convolution
↓
ReLU
↓
Pooling

VGG models are useful for learning CNN architecture, but they are generally not the first choice when efficiency is important.

## 3. ResNet50

ResNet stands for:

Residual Network

ResNet introduced an important idea called:

Residual Connections / Skip Connections

Instead of forcing every layer to learn a completely new representation, information can skip some layers through a shortcut connection.

Conceptually:

Input
  │
  ├───────────────┐
  ↓               │
Layer 1           │
  ↓               │
Layer 2           │
  ↓               │
  └────── + <─────┘
         ↓
       Output

This helps very deep networks train effectively.

ResNet50 is widely used for Transfer Learning.

## 4. MobileNetV2

MobileNetV2 is designed to be lightweight and computationally efficient.

It is useful when:

- We have limited computational resources
- We want faster inference
- We are building mobile applications
- We want a relatively small model

For our Cat vs Dog Transfer Learning project, MobileNetV2 is a very good model to start with.

The basic workflow will be:

MobileNetV2
      ↓
Remove original classifier
      ↓
Freeze pretrained layers
      ↓
Add our classifier
      ↓
Cat / Dog prediction

## 5. InceptionV3

InceptionV3 uses special modules that process information using different convolution operations.

Conceptually:

                 Input
                   ↓
        ┌──────────┼──────────┐
        ↓          ↓          ↓
    Conv 1×1    Conv 3×3    Conv 5×5
        ↓          ↓          ↓
        └──────────┼──────────┘
                   ↓
                Combine
                   ↓
                Output

The architecture allows the network to capture information at different scales.

## 6. EfficientNet

EfficientNet is designed to achieve a strong balance between:

- Accuracy
- Model size
- Computational cost

EfficientNet has multiple versions such as:

EfficientNetB0
EfficientNetB1
EfficientNetB2
...
EfficientNetB7

Generally, larger versions require more computational resources.

EfficientNet is another strong option for Transfer Learning.

## Comparing Pretrained Models

| Model | Main Idea | Size/Cost | Good For |
|---|---|---|---|
| VGG16 | Simple deep CNN | Large | Learning / classic CNN |
| VGG19 | Deeper VGG | Large | Learning / feature extraction |
| ResNet50 | Skip connections | Medium | General computer vision |
| MobileNetV2 | Lightweight CNN | Small | Fast/mobile applications |
| InceptionV3 | Multi-scale processing | Medium/Large | Image classification |
| EfficientNet | Efficient scaling | Varies | Strong accuracy/efficiency |

There is no single best model for every problem.

The best model depends on:

- Dataset
- Hardware
- Speed requirements
- Accuracy requirements
- Model size

## Loading a Pretrained Model

We will use TensorFlow/Keras to load pretrained models.

Keras provides many pretrained architectures through:

`tf.keras.applications`

```python
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
```

## Loading MobileNetV2

We can load MobileNetV2 with ImageNet pretrained weights.

`include_top=False` means we remove the original classification head.

We do this because the original classifier was designed for the original ImageNet classes.

Our problem is different.

For example:

ImageNet:
1000 classes

Our project:
2 classes → Cat and Dog

Therefore, we keep the CNN feature extractor and create our own classifier.

```python
base_model = tf.keras.applications.MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

print("MobileNetV2 loaded successfully!")
```

## Understanding the Arguments

We used:

weights="imagenet"

This tells Keras to load weights pretrained on ImageNet.

include_top=False

This removes the original classification layers.

input_shape=(224, 224, 3)

This means:

224 → image height

224 → image width

3 → RGB channels

Therefore, our input image has the shape:

(224, 224, 3)

```python
base_model.summary()
```

## Inspecting the Model

We can check how many layers the pretrained model contains.

Each layer may contain:

- Weights
- Biases
- Convolution operations
- Batch normalization
- Activation functions

These layers together form the pretrained CNN.

```python
print("Number of layers:", len(base_model.layers))
```

## Inspecting Individual Layers

We can look at the layers one by one.

This helps us understand what exists inside a pretrained CNN.

```python
for i, layer in enumerate(base_model.layers[:20]):
    print(i, layer.name, layer.__class__.__name__)
```

## Trainable Layers

By default, the pretrained model can be trainable.

For Feature Extraction, we usually freeze the pretrained model.

We can do this using:

`base_model.trainable = False`

After freezing:

The pretrained weights will not be updated during training.

```python
base_model.trainable = False

print("Base model trainable:", base_model.trainable)
```

## Trainable vs Non-Trainable Parameters

When we freeze the pretrained CNN:

Pretrained CNN
     ↓
Weights frozen
     ↓
No weight updates

Later, when we add our own classifier:

New Classifier
     ↓
Trainable
     ↓
Weights updated

This is the foundation of Feature Extraction.

```python
trainable_count = sum(
    tf.keras.backend.count_params(w)
    for w in base_model.trainable_weights
)

non_trainable_count = sum(
    tf.keras.backend.count_params(w)
    for w in base_model.non_trainable_weights
)

print("Trainable parameters:", trainable_count)
print("Non-trainable parameters:", non_trainable_count)
```
