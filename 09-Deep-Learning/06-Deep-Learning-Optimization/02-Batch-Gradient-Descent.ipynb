# Batch Gradient Descent

Batch Gradient Descent is an optimization algorithm used to train machine learning and deep learning models.

It calculates the gradient using the **entire training dataset** before updating the model's weights.

The basic process is:

**All Training Data**
↓
**Forward Pass**
↓
**Calculate Loss**
↓
**Calculate Gradients**
↓
**Update Weights**
↓
**Repeat**

## What is Gradient Descent?

Gradient Descent is an optimization algorithm.

Its goal is to minimize the loss function.

The basic weight update equation is:

$$
W_{new} = W_{old} - \eta \nabla_W L
$$

Where:

- $W$ = model weights
- $\eta$ = learning rate
- $\nabla_W L$ = gradient of the loss with respect to the weights
- $L$ = loss

The gradient tells us which direction the loss is increasing.

Therefore, we move in the opposite direction to reduce the loss.

## Why is it called Batch Gradient Descent?

Suppose we have:

$$
1000
$$

training examples.

In Batch Gradient Descent, we use **all 1000 examples** to calculate the gradient.

Then we perform one weight update.

So:

$$
1000\ samples
\rightarrow
1\ gradient
\rightarrow
1\ weight\ update
$$

The complete training dataset is called one batch for that update.

## Batch Gradient Descent Formula

Suppose we have $m$ training examples.

The average gradient is:

$$
\nabla_W L
=
\frac{1}{m}
\sum_{i=1}^{m}
\nabla_W L_i
$$

Then the weights are updated:

$$
\boxed{
W_{new}
=
W_{old}
-
\eta
\frac{1}{m}
\sum_{i=1}^{m}
\nabla_W L_i
}
$$

So Batch Gradient Descent calculates the gradient from every training example and averages them before updating the weights.

```python
gradients = [2, 4, 6]

batch_gradient = sum(gradients) / len(gradients)

weight = 5
learning_rate = 0.1

new_weight = weight - learning_rate * batch_gradient

print("Batch gradient:", batch_gradient)
print("Old weight:", weight)
print("New weight:", new_weight)
```

## Batch Gradient Descent Workflow

Suppose we have 1000 training samples.

### Step 1 — Take the complete dataset

$$
X_{train}, y_{train}
$$

### Step 2 — Forward propagation

$$
\hat{y}=model(X_{train})
$$

### Step 3 — Calculate loss

$$
L=Loss(y,\hat{y})
$$

### Step 4 — Backpropagation

Calculate:

$$
\nabla_W L
$$

### Step 5 — Update weights

$$
W=W-\eta\nabla_W L
$$

### Step 6 — Repeat for the next epoch

The complete dataset is processed again.

This continues until the model converges or training stops.

## Epoch vs Batch

These two terms are very important.

### Epoch

One epoch means the model has gone through the **entire training dataset once**.

### Batch

A batch is the group of training samples used for one gradient calculation/update.

In pure Batch Gradient Descent:

$$
Batch\ Size = Dataset\ Size
$$

Example:

Dataset:

$$
1000\ samples
$$

Batch size:

$$
1000
$$

Then:

**1 epoch = 1 batch = 1 weight update**

## Three Types of Gradient Descent

There are three common approaches.

### 1. Batch Gradient Descent

Uses the entire dataset.

$$
Batch\ Size = N
$$

For 1000 samples:

$$
1000\ samples
\rightarrow
1\ update
$$

---

### 2. Stochastic Gradient Descent (SGD)

Uses one sample at a time.

$$
Batch\ Size = 1
$$

For 1000 samples:

$$
1000\ samples
\rightarrow
1000\ updates
$$

---

### 3. Mini-Batch Gradient Descent

Uses a small group of samples.

For example:

$$
Batch\ Size = 32
$$

For 1000 samples:

Approximately:

$$
\frac{1000}{32}\approx32
$$

updates per epoch.

Modern deep learning normally uses **Mini-Batch Gradient Descent**.

## Comparison

| Method | Batch Size | Updates per Epoch | Speed |
|---|---:|---:|---|
| Batch Gradient Descent | Entire dataset | 1 | Slow for large data |
| SGD | 1 | N | Noisy |
| Mini-Batch | Small batch | N / batch size | Usually efficient |

### Important

When training CNNs, we normally use **mini-batches**, such as:

$$
32,\ 64,\ 128
$$

rather than using the entire image dataset at once.

```python
import numpy as np

# Training data
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)

# Initial parameters
w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000

n = len(X)

for epoch in range(epochs):

    # Forward pass
    y_pred = w * X + b

    # Error
    error = y_pred - y

    # Mean Squared Error
    loss = np.mean(error ** 2)

    # Gradients using the complete dataset
    dw = (2 / n) * np.sum(error * X)
    db = (2 / n) * np.sum(error)

    # Update weights
    w = w - learning_rate * dw
    b = b - learning_rate * db

    if epoch % 100 == 0:
        print(
            f"Epoch {epoch}: "
            f"Loss={loss:.4f}, "
            f"w={w:.4f}, "
            f"b={b:.4f}"
        )
```
