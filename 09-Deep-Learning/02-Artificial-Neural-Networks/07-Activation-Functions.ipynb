# Activation Functions

An activation function is a mathematical function applied to the
weighted sum of a neuron.

The basic neuron equation is:

z = Σ(wᵢxᵢ) + b

The activation function then transforms z:

a = f(z)

Where:

- z = Weighted Sum + Bias
- f = Activation Function
- a = Neuron Output

Activation functions introduce non-linearity into neural networks.

# Main Activation Functions

Important activation functions include:

1. Step Function
2. Sigmoid
3. Tanh
4. ReLU
5. Leaky ReLU
6. ELU
7. Softmax

Different activation functions are useful for different situations.

## Step Function

The Step Function produces a binary output.

It can be defined as:

f(z) = 1, if z >= 0

f(z) = 0, if z < 0

Therefore:

z < 0  →  0

z >= 0 →  1

The Step Function was commonly associated with the original
Perceptron.

```python
def step_function(z):
    if z >= 0:
        return 1
    else:
        return 0


print(step_function(-2))## Limitation of Step Function

The Step Function has an important limitation.

It is not differentiable at the threshold and its derivative is
zero almost everywhere.

Neural networks are commonly trained using gradient-based methods
such as backpropagation.

Therefore, the Step Function is generally not used in hidden layers
of modern neural networks.
print(step_function(3))
```

## Sigmoid Function

The Sigmoid function converts any real-valued number into a value
between 0 and 1.

Formula:

σ(z) = 1 / (1 + e⁻ᶻ)

Range:

0 < σ(z) < 1

Important values:

z → -∞  → 0

z = 0   → 0.5

z → +∞  → 1

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))


print(sigmoid(-5))
print(sigmoid(0))
print(sigmoid(5))
```

## Tanh Function

Tanh stands for Hyperbolic Tangent.

Formula:

tanh(z) = (eᶻ - e⁻ᶻ) / (eᶻ + e⁻ᶻ)

Range:

-1 < tanh(z) < 1

Important values:

z → -∞ → -1

z = 0 → 0

z → +∞ → 1

```python
def tanh(z):
    return np.tanh(z)


print(tanh(-2))
print(tanh(0))
print(tanh(2))
```

## ReLU

ReLU stands for:

Rectified Linear Unit

It is one of the most commonly used activation functions in
modern neural networks.

Formula:

ReLU(z) = max(0, z)

Therefore:

z < 0  →  0

z >= 0 →  z

```python
def relu(z):
    return np.maximum(0, z)


print(relu(-5))
print(relu(0))
print(relu(5))
```

## Leaky ReLU

Leaky ReLU is a modified version of ReLU.

Formula:

LeakyReLU(z) = z, if z >= 0

LeakyReLU(z) = αz, if z < 0

where α is a small positive value such as:

α = 0.01

```python
def leaky_relu(z, alpha=0.01):
    return np.where(z >= 0, z, alpha * z)


print(leaky_relu(-5))
print(leaky_relu(5))
```

## Softmax Function

Softmax is commonly used in the output layer of a multiclass
classification neural network.

For multiple class scores:

z₁, z₂, ..., zₙ

Softmax converts them into probabilities.

Formula:

softmax(zᵢ) = eᶻⁱ / Σeᶻʲ

The resulting probabilities:

- Are between 0 and 1.
- Add up to 1.

Example:

Class A → 0.10
Class B → 0.70
Class C → 0.20

Total:

0.10 + 0.70 + 0.20 = 1.00

```python
def softmax(z):
    exp_values = np.exp(z - np.max(z))
    return exp_values / np.sum(exp_values)


scores = np.array([2.0, 4.0, 1.0])

probabilities = softmax(scores)

print(probabilities)
print("Sum:", np.sum(probabilities))
```

## Activation Function Comparison

| Function | Output Range | Common Use |
|---|---|---|
| Step | 0 or 1 | Perceptron |
| Sigmoid | 0 to 1 | Binary classification output |
| Tanh | -1 to 1 | Some hidden layers / older architectures |
| ReLU | 0 to ∞ | Common hidden-layer activation |
| Leaky ReLU | -∞ to ∞ | Alternative to ReLU |
| ELU | -α to ∞ | Alternative activation |
| Softmax | 0 to 1 | Multiclass classification output |

There is no single activation function that is best for every
problem.

## Which Activation Function Should I Use?

### Hidden Layers

A common default choice is:

ReLU

Other possibilities include:

- Leaky ReLU
- ELU
- GELU
- Tanh

### Binary Classification Output

Usually:

Sigmoid

Output:

0 to 1

### Multiclass Classification Output

Usually:

Softmax

Output:

A probability for each class.

### Regression Output

Usually:

Linear activation

The output is not restricted to a fixed range.
