# Why Do We Use Recurrent Neural Networks (RNNs)?

Recurrent Neural Networks (RNNs) are used when the **order and relationship between data points are important**.

Traditional neural networks are very good at learning from independent inputs, but many real-world problems contain sequences where previous information can affect the current prediction.

RNNs are designed to process these sequences while maintaining information from previous time steps.

## Why Not Use a Traditional Neural Network?

A traditional feed-forward neural network generally processes its inputs without maintaining a recurrent hidden state from one time step to the next.

For sequential problems, this can be limiting.

Suppose we have:

```text
I → love → machine → learning

## RNN Provides a Form of Memory

One of the main reasons we use RNNs is their ability to maintain information from previous time steps.

At each time step, the RNN receives:

- Current input
- Previous hidden state

It then produces a new hidden state.

Conceptually:

```text
Current Input + Previous Information
              ↓
             RNN
              ↓
       New Hidden State

## Main Reasons to Use RNNs

We use RNNs mainly because they are designed for sequential data.

### 1. Sequential Processing

RNNs process data one time step at a time.

### 2. Hidden-State Memory

RNNs maintain a hidden state that carries information from previous time steps.

### 3. Order Awareness

RNNs preserve the order of observations in a sequence.

### 4. Variable-Length Sequences

RNN architectures can work with sequences containing different numbers of time steps.

### 5. Sequence Prediction

RNNs can use previous information to help predict current or future outputs.

### 6. Useful for Many Sequential Problems

RNNs can be applied to:

- Text
- Speech
- Time series
- Sensor data
- Sequence classification
- Sequence generation
