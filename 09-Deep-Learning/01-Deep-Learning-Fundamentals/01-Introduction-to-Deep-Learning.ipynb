# Introduction to Deep Learning

Deep Learning is a subfield of Machine Learning that uses
artificial neural networks with multiple layers to learn
complex patterns from data.

Unlike traditional machine learning, where we often need to
manually engineer important features, deep learning models can
automatically learn useful representations and features from
raw or minimally processed data.

Deep Learning is especially powerful for:

- Image recognition
- Computer vision
- Natural language processing
- Speech recognition
- Time-series data
- Recommendation systems
- Generative AI
- Large Language Models (LLMs)

## Artificial Intelligence → Machine Learning → Deep Learning

Artificial Intelligence (AI) is the broad field of creating
systems that can perform tasks that normally require human
intelligence.

Machine Learning (ML) is a subset of AI where algorithms learn
patterns from data instead of being explicitly programmed for
every rule.

Deep Learning (DL) is a subset of Machine Learning that uses
multi-layer artificial neural networks to learn complex patterns.

## Artificial Neural Network

An Artificial Neural Network (ANN) is a computational model
inspired by the structure of biological neural networks.

An ANN consists of interconnected artificial neurons.

A basic neural network contains:

- Input Layer
- Hidden Layer(s)
- Output Layer

The neurons use weights, biases, and activation functions
to transform inputs and produce outputs.

## How Deep Learning Works

A neural network learns by repeatedly performing the following
steps:

1. Receive input data.
2. Multiply inputs by weights.
3. Add biases.
4. Apply activation functions.
5. Produce predictions.
6. Calculate the loss.
7. Use backpropagation to calculate gradients.
8. Update the weights.
9. Repeat the process for many iterations.

The objective is to reduce the loss and improve predictions.

## Why is it Called Deep Learning?

The word "deep" refers to the presence of multiple hidden layers
in a neural network.

A network with one hidden layer can be considered a neural network,
while a network with many hidden layers is commonly referred to
as a deep neural network.

Example:

Input
  ↓
Hidden Layer 1
  ↓
Hidden Layer 2
  ↓
Hidden Layer 3
  ↓
Hidden Layer 4
  ↓
Output

More layers allow the network to learn increasingly complex
representations of the data.
