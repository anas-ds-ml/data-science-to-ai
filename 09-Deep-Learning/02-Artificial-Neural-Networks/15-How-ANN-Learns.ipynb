# How Does an Artificial Neural Network Learn?

An Artificial Neural Network (ANN) learns by adjusting its
weights and biases so that its predictions become closer to
the actual target values.

The learning process is based on:

1. Forward Propagation
2. Loss Calculation
3. Backpropagation
4. Gradient Calculation
5. Weight and Bias Updates
6. Repetition

# Basic Idea of ANN Learning

Suppose:

Actual Value = 1

Initial Prediction = 0.2

The prediction is not good.

The neural network calculates the loss and determines how its
weights and biases should change.

After updating:

Prediction = 0.4

After more training:

Prediction = 0.7

Later:

Prediction = 0.95

The network gradually learns parameters that produce better
predictions.

# ANN Learning Cycle

The complete learning cycle is:

Input Data
    ↓
Forward Propagation
    ↓
Prediction
    ↓
Loss Function
    ↓
Loss
    ↓
Backpropagation
    ↓
Gradients
    ↓
Optimizer
    ↓
Weight & Bias Updates
    ↓
Next Prediction
    ↓
Repeat
