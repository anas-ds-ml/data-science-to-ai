# Padding and Stride in CNN

In Convolutional Neural Networks (CNNs), **Padding** and **Stride**
control how a filter moves over an image and how large the output
feature map will be.

### Padding
Padding adds extra pixels around the border of an image.

### Stride
Stride determines how many pixels the filter moves at each step.

Both are important because they control the spatial dimensions of
the feature map produced by convolution.

## 1. What is Padding?

Padding means adding extra pixels around the border of an image
before applying convolution.

The most common padding value is zero.

For example, a 3 × 3 image:

    1 2 3
    4 5 6
    7 8 9

with padding = 1 becomes:

    0 0 0 0 0
    0 1 2 3 0
    0 4 5 6 0
    0 7 8 9 0
    0 0 0 0 0

Padding helps preserve information near the edges and can prevent
the feature map from becoming smaller too quickly.

## 2. Types of Padding

There are two important padding modes in CNNs:

### 1. Valid Padding

No padding is added.

```python
padding="valid"

````python
import tensorflow as tf

image = tf.random.normal((1, 5, 5, 1))

conv_valid = tf.keras.layers.Conv2D(
    filters=1,
    kernel_size=(3, 3),
    strides=(1, 1),
    padding="valid"
)

output_valid = conv_valid(image)

print("Input shape :", image.shape)
print("Output shape:", output_valid.shape)
````

````python
conv_same = tf.keras.layers.Conv2D(
    filters=1,
    kernel_size=(3, 3),
    strides=(1, 1),
    padding="same"
)

output_same = conv_same(image)

print("Input shape :", image.shape)
print("Output shape:", output_same.shape)
````

## 3. What is Stride?

Stride determines how many pixels the convolution filter moves
after each operation.

### Stride = 1

The filter moves one pixel at a time.

    → → →

This produces a larger and more detailed feature map.

### Stride = 2

The filter moves two pixels at a time.

    →   →   →

This reduces the spatial size of the feature map.

In TensorFlow:

```python
strides=(1, 1)

````python
conv_stride_1 = tf.keras.layers.Conv2D(
    filters=1,
    kernel_size=(3, 3),
    strides=(1, 1),
    padding="valid"
)

output_stride_1 = conv_stride_1(image)

print("Input shape :", image.shape)
print("Output shape:", output_stride_1.shape)
````

````python
conv_stride_2 = tf.keras.layers.Conv2D(
    filters=1,
    kernel_size=(3, 3),
    strides=(2, 2),
    padding="valid"
)

output_stride_2 = conv_stride_2(image)

print("Input shape :", image.shape)
print("Output shape:", output_stride_2.shape)
````

## 4. Convolution Output Size Formula

The output spatial size of a convolution can be calculated using:

Output Size = floor((N + 2P - K) / S) + 1

Where:

- N = Input size
- P = Padding
- K = Kernel size
- S = Stride

This formula helps us calculate the height and width of the
resulting feature map.

````python
image = tf.random.normal((1, 64, 64, 3))

configs = [
    ("valid, stride 1", "valid", (1, 1)),
    ("same, stride 1", "same", (1, 1)),
    ("valid, stride 2", "valid", (2, 2)),
    ("same, stride 2", "same", (2, 2))
]

for name, padding, stride in configs:

    conv = tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        strides=stride,
        padding=padding,
        activation="relu"
    )

    output = conv(image)

    print(name)
    print("Output shape:", output.shape)
    print("-" * 40)
````
