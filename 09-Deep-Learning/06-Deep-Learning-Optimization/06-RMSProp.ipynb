# RMSProp Optimization

RMSProp stands for:

    Root Mean Square Propagation

RMSProp is an optimization algorithm used to train
neural networks and deep learning models.

RMSProp improves Gradient Descent by automatically
adjusting the learning rate for each parameter.

The main idea is:

    Large historical gradients
            ↓
    smaller future updates

    Small historical gradients
            ↓
    relatively larger updates

## Why Do We Need RMSProp?

Suppose a neural network has two weights:

    W1
    W2

Imagine their gradients behave differently:

    W1 → gradients are very large
    W2 → gradients are very small

Using the same learning rate for both can be inefficient.

For example:

    learning_rate = 0.01

A large gradient can cause:

    very large updates

while a small gradient can cause:

    very small updates

RMSProp adapts the update for each parameter.

## Main Idea of RMSProp

RMSProp keeps a moving average of the
squared gradients.

It remembers:

    How large have the recent gradients been?

Then it uses this information to scale
the current gradient.

So RMSProp automatically adjusts the step size
for each parameter.

## Understanding the Formula

First:

    s = rho × previous_s
        +
        (1-rho) × gradient²

This keeps track of the recent magnitude
of the gradients.

Then:

    effective_step =
        learning_rate / sqrt(s + epsilon)

Finally:

    weight_update =
        effective_step × gradient

Therefore, if the historical squared gradients
become large:

    sqrt(s) becomes large
           ↓
    effective step becomes smaller

If the historical gradients are small:

    sqrt(s) becomes smaller
           ↓
    effective step becomes larger

```python
import numpy as np

w = 5.0

learning_rate = 0.1
rho = 0.9
epsilon = 1e-8

s = 0.0

gradients = [2, 2, 2, 2, 2]

for step, gradient in enumerate(gradients, start=1):

    # Moving average of squared gradients
    s = rho * s + (1 - rho) * gradient**2

    # RMSProp update
    update = learning_rate * gradient / (
        np.sqrt(s) + epsilon
    )

    # Update weight
    w = w - update

    print(
        f"Step {step}: "
        f"Gradient={gradient}, "
        f"s={s:.4f}, "
        f"Update={update:.4f}, "
        f"Weight={w:.4f}"
    )
```
