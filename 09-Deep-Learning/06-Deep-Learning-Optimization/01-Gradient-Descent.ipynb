# Gradient Descent

Gradient Descent is an optimization algorithm used to minimize
the loss function of a machine learning or neural network model.

The main goal is:

Minimize Loss

It does this by repeatedly updating the model's parameters
(weights and biases) in the direction that reduces the loss.

# Basic Idea

Imagine that the loss function is a mountain.

We are standing somewhere on the mountain and want to reach
the lowest point.

The gradient tells us the direction in which the loss increases.

Gradient Descent moves in the opposite direction.

Therefore:

Gradient
    ↓
Direction of increasing loss

Negative Gradient
    ↓
Direction toward lower loss

# Gradient Descent Update Rule

For a weight:

w_new = w_old - η(∂L/∂w)

For a bias:

b_new = b_old - η(∂L/∂b)

Where:

w = Weight
b = Bias
L = Loss
η = Learning Rate
∂L/∂w = Gradient of loss with respect to weight
∂L/∂b = Gradient of loss with respect to bias

The minus sign is important because we want to move in the
opposite direction of the gradient.

# What is a Gradient?

A gradient tells us how the loss changes when a parameter changes.

For example:

∂L/∂w = 5

This means that locally, increasing w tends to increase the loss.

Gradient Descent therefore moves w in the opposite direction.

If:

∂L/∂w = -5

Gradient Descent moves w in the positive direction.

Therefore:

Positive Gradient → Parameter decreases

Negative Gradient → Parameter increases

# Learning Rate

The learning rate determines how large the update step will be.

It is represented by:

η

Example:

η = 0.01

Gradient = 5

Update:

η × Gradient

= 0.01 × 5

= 0.05

Therefore:

w_new = w_old - 0.05

```python
w = 2.0
gradient = 4.0
learning_rate = 0.1

w_new = w - learning_rate * gradient

print("Old Weight:", w)
print("Gradient:", gradient)
print("New Weight:", w_new)
```

# Gradient Descent on a Simple Function

Consider:

f(w) = w²

The derivative is:

f'(w) = 2w

The minimum of this function occurs at:

w = 0

Gradient Descent will try to move w toward zero.

```python
w = 5.0
learning_rate = 0.1

for step in range(20):
    gradient = 2 * w
    w = w - learning_rate * gradient
    
    print(f"Step {step + 1}: w = {w:.4f}")
```

# Important Distinction

Backpropagation:

Calculates the gradients.

Gradient Descent:

Uses the gradients to update parameters.

Therefore:

Backpropagation ≠ Gradient Descent

They work together.

Backpropagation:
"What is the gradient?"

Gradient Descent:
"Use that gradient to update the parameter."
