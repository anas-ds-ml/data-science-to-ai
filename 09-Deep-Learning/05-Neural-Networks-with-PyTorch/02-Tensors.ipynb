# PyTorch Tensors

A tensor is the main data structure used by PyTorch.

A tensor is a multi-dimensional array that can store numbers.

Tensors are used to represent:

- Input data
- Features
- Labels
- Weights
- Biases
- Images
- Audio
- Text representations
- Neural network outputs

PyTorch performs mathematical operations on tensors
and can move tensors to the GPU for faster computation.

```python
import torch

print(torch.__version__)
```

## Scalar

A scalar is a tensor containing a single value.

Example:

5

A scalar has:

- 0 dimensions
- Shape: []

```python
x = torch.tensor(5)

print(x)
print("Shape:", x.shape)
print("Dimensions:", x.ndim)
```

## Vector

A vector is a one-dimensional tensor.

Example:

[10, 20, 30, 40]

It has one dimension.

```python
x = torch.tensor([10, 20, 30, 40])

print(x)
print("Shape:", x.shape)
print("Dimensions:", x.ndim)
```

## Matrix

A matrix is a two-dimensional tensor.

It contains rows and columns.

Example:

[[1, 2, 3],
 [4, 5, 6]]

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(x)

print("Shape:", x.shape)
print("Dimensions:", x.ndim)
```

## 3D Tensor

A 3D tensor contains multiple matrices.

Example:

[
    [
        [1, 2],
        [3, 4]
    ],

    [
        [5, 6],
        [7, 8]
    ]
]

Shape:

(2, 2, 2)

```python
x = torch.tensor([
    [
        [1, 2],
        [3, 4]
    ],
    
    [
        [5, 6],
        [7, 8]
    ]
])

print(x)

print("Shape:", x.shape)
print("Dimensions:", x.ndim)
```

## Tensor Shape

Shape tells us the size of a tensor along each dimension.

For example:

X.shape = (1000, 10)

This means:

1000 → number of samples
10   → number of features

This is exactly the type of tensor used as input
to an ANN.

```python
X = torch.randn(1000, 10)

print(X.shape)
```

## Tensor Data Types

PyTorch tensors can contain different data types.

Common types:

- torch.float32
- torch.float64
- torch.int32
- torch.int64
- torch.bool

For neural network calculations, float32 is commonly used.

```python
x = torch.tensor([1.0, 2.0, 3.0])

print(x)
print(x.dtype)
```

```python
# Tensor from values
x = torch.tensor([1, 2, 3])

# Tensor filled with zeros
zeros = torch.zeros(2, 3)

# Tensor filled with ones
ones = torch.ones(2, 3)

# Random tensor
random = torch.rand(2, 3)

print(x)
print(zeros)
print(ones)
print(random)
```
