# Partial Derivatives

In deep learning, a model usually has many variables.

For example:

$$
z = wx + b
$$

Here we have multiple variables:

- $w$ = weight
- $x$ = input
- $b$ = bias

A **partial derivative** tells us how the output changes with respect to **one variable while treating the other variables as constants**.

This is extremely important in neural networks because we need to know how the loss changes with respect to each individual weight and bias.

## Ordinary Derivative vs Partial Derivative

### Ordinary Derivative

If a function has one variable:

$$
y=x^2
$$

we calculate:

$$
\frac{dy}{dx}=2x
$$

This is an ordinary derivative.

---

### Partial Derivative

If a function has multiple variables:

$$
z=x^2+y^2
$$

we can calculate the partial derivative with respect to $x$:

$$
\frac{\partial z}{\partial x}=2x
$$

while treating $y$ as a constant.

We can also calculate the partial derivative with respect to $y$:

$$
\frac{\partial z}{\partial y}=2y
$$

The symbol:

$$
\partial
$$

is used for partial derivatives.

```python
x = 3
y = 4

dz_dx = 2 * x
dz_dy = 2 * y

print("∂z/∂x =", dz_dx)
print("∂z/∂y =", dz_dy)
```

## The Most Important Rule

When calculating a partial derivative with respect to one variable:

> Treat all other variables as constants.

For example:

$$
z=x^2+3xy+y^2
$$

If we calculate:

$$
\frac{\partial z}{\partial x}
$$

then $y$ is treated as a constant.

Therefore:

$$
\frac{\partial x^2}{\partial x}=2x
$$

and:

$$
\frac{\partial 3xy}{\partial x}=3y
$$

and:

$$
\frac{\partial y^2}{\partial x}=0
$$

Therefore:

$$
\boxed{
\frac{\partial z}{\partial x}=2x+3y
}
$$

```python
x = 2
y = 3

partial_x = 2*x + 3*y

print("∂z/∂x =", partial_x)
```

## Partial Derivatives in a Neural Network

Consider a simple neuron:

$$
z=wx+b
$$

There are three variables:

- $w$ = weight
- $x$ = input
- $b$ = bias

We can calculate the partial derivative with respect to the weight:

$$
\frac{\partial z}{\partial w}
$$

Since $x$ is treated as a constant:

$$
\boxed{
\frac{\partial z}{\partial w}=x
}
$$

Similarly:

$$
\boxed{
\frac{\partial z}{\partial b}=1
}
$$

And:

$$
\boxed{
\frac{\partial z}{\partial x}=w
}
$$

```python
x = 2
w = 3

dz_dw = x
dz_db = 1
dz_dx = w

print("∂z/∂w =", dz_dw)
print("∂z/∂b =", dz_db)
print("∂z/∂x =", dz_dx)
```

## Why Do Neural Networks Need Partial Derivatives?

A neural network can contain thousands or millions of parameters.

For example:

$$
W=[w_1,w_2,w_3]
$$

The network needs to know how the loss changes with respect to every weight:

$$
\frac{\partial L}{\partial w_1}
$$

$$
\frac{\partial L}{\partial w_2}
$$

$$
\frac{\partial L}{\partial w_3}
$$

These partial derivatives are combined to form the gradient:

$$
\nabla_W L =
\begin{bmatrix}
\frac{\partial L}{\partial w_1}\\
\frac{\partial L}{\partial w_2}\\
\frac{\partial L}{\partial w_3}
\end{bmatrix}
$$

Therefore:

**Partial derivatives → Gradient → Weight updates**

## Partial Derivative of Loss

Suppose:

$$
z=wx+b
$$

and our loss is:

$$
L=(z-y)^2
$$

We want to find:

$$
\frac{\partial L}{\partial w}
$$

Because $L$ depends on $w$ through $z$, we use the chain rule:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial z}
\frac{\partial z}{\partial w}
$$

First:

$$
L=(z-y)^2
$$

Therefore:

$$
\frac{\partial L}{\partial z}=2(z-y)
$$

And:

$$
z=wx+b
$$

so:

$$
\frac{\partial z}{\partial w}=x
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}
=
2(z-y)x
}
$$

This is the gradient of the loss with respect to the weight.

```python
x = 2
w = 3
b = 1
y = 10

# Forward pass
z = w * x + b

# Loss
loss = (z - y) ** 2

# Partial derivative dL/dw
dL_dw = 2 * (z - y) * x

print("z =", z)
print("Loss =", loss)
print("∂L/∂w =", dL_dw)
```

```python
learning_rate = 0.1

new_w = w - learning_rate * dL_dw

print("Old weight:", w)
print("Gradient:", dL_dw)
print("New weight:", new_w)
```
