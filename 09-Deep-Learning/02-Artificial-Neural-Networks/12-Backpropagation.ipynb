# Backpropagation

Backpropagation is an algorithm used to calculate the gradients
of the loss function with respect to the weights and biases of
a neural network.

These gradients are then used by an optimizer to update the
network's parameters.

In simple words:

Backpropagation tells the neural network:

"Prediction wrong hai, ab weights aur biases ko kis direction
mein change karna hai?"

# Why Do We Need Backpropagation?

A neural network contains many parameters.

For example:

Weights:
w₁, w₂, w₃, ...

Biases:
b₁, b₂, b₃, ...

We need to determine how each parameter affects the loss.

Backpropagation calculates:

∂L/∂w

and

∂L/∂b

These values are called gradients.

The gradients tell us how the loss changes when a parameter
changes.

# Complete Training Process

The neural network learns using the following process:

1. Initialize weights and biases
2. Forward propagation
3. Generate prediction
4. Calculate loss
5. Backpropagation
6. Calculate gradients
7. Update weights and biases
8. Repeat

Flow:

Input
  ↓
Forward Propagation
  ↓
Prediction
  ↓
Loss
  ↓
Backpropagation
  ↓
Gradients
  ↓
Optimizer
  ↓
Updated Parameters
  ↓
Repeat

# What is a Gradient?

A gradient tells us how much the loss changes with respect to
a parameter.

For a weight:

∂L/∂w

For a bias:

∂L/∂b

If the gradient is positive:

Increasing the parameter tends to increase the loss locally.

If the gradient is negative:

Increasing the parameter tends to decrease the loss locally.

The optimizer uses this information to update the parameters.

# Gradient Descent Update Rule

The basic weight update rule is:

w_new = w_old - η(∂L/∂w)

The bias update rule is:

b_new = b_old - η(∂L/∂b)

Where:

w = Weight
b = Bias
L = Loss
η = Learning Rate
∂L/∂w = Gradient with respect to weight
∂L/∂b = Gradient with respect to bias

# Learning Rate

The learning rate controls how large each parameter update is.

Let:

η = 0.01

If:

Gradient = 2

Then:

Update = 0.01 × 2

Update = 0.02

The weight is updated as:

w_new = w_old - 0.02

# Chain Rule

Backpropagation relies heavily on the mathematical Chain Rule.

The Chain Rule allows us to calculate how a change in one
variable affects the final loss through multiple operations.

Suppose:

L depends on y

y depends on z

z depends on w

Then:

∂L/∂w =
(∂L/∂y)
×
(∂y/∂z)
×
(∂z/∂w)

This allows the gradient to be propagated backward through
the network.

# Why is it Called Backpropagation?

The error is calculated at the output.

Then the gradient information is propagated backward through
the network.

Forward:

Input → Hidden → Output → Loss

Backward:

Loss → Output → Hidden → Input

Therefore:

"Back" = Moving backward through the network

"Propagation" = Propagating gradient information

```python
x = 2
w = 0.5
b = 1
y = 5

learning_rate = 0.01

# Forward propagation
y_pred = w * x + b

# Loss
loss = (y - y_pred) ** 2

# Gradient of loss with respect to weight
gradient_w = 2 * (y_pred - y) * x

# Update weight
w_new = w - learning_rate * gradient_w

print("Prediction:", y_pred)
print("Loss:", loss)
print("Gradient:", gradient_w)
print("Old Weight:", w)
print("New Weight:", w_new)
```
