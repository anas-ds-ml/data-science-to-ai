# Mini-Batch Gradient Descent

Mini-Batch Gradient Descent is an optimization algorithm used to train
Machine Learning and Deep Learning models.

It combines the advantages of:

- Batch Gradient Descent
- Stochastic Gradient Descent

Instead of using the entire dataset or only one sample, Mini-Batch Gradient
Descent divides the dataset into small groups called **mini-batches**.

Example:

Dataset = 1000 samples

Batch Gradient Descent:
    1000 samples → 1 update

Stochastic Gradient Descent:
    1 sample → 1 update

Mini-Batch Gradient Descent:
    32 samples → 1 update

## What is Mini-Batch Gradient Descent?

Mini-Batch Gradient Descent divides the training dataset into small batches.

For example, suppose we have:

    1000 training samples

and choose:

    batch_size = 32

The model processes approximately:

    32 samples → calculate loss → backpropagation → update weights

Then:

    next 32 samples → calculate loss → backpropagation → update weights

This continues until the complete dataset has been processed.

After the complete dataset is processed, one **epoch** is completed.

## Batch Size

Batch size is the number of training examples processed before updating
the model's weights.

Common batch sizes are:

    16
    32
    64
    128

For example:

    batch_size = 32

means:

    32 samples → one gradient calculation → one weight update

## Batch GD vs SGD vs Mini-Batch GD

Suppose our dataset contains 1000 samples.

### Batch Gradient Descent

Batch size:

    1000

Updates per epoch:

    1

### Stochastic Gradient Descent

Batch size:

    1

Updates per epoch:

    1000

### Mini-Batch Gradient Descent

Batch size:

    32

Updates per epoch:

    approximately 32

Therefore:

Batch GD:
    Large batch → fewer updates

SGD:
    One sample → many updates

Mini-Batch GD:
    Small batch → balanced number of updates

## Simple Example

Suppose we have 8 training samples:

    [1, 2, 3, 4, 5, 6, 7, 8]

Choose:

    batch_size = 2

The dataset becomes:

    Batch 1 → [1, 2]
    Batch 2 → [3, 4]
    Batch 3 → [5, 6]
    Batch 4 → [7, 8]

For every mini-batch:

    Forward Pass
         ↓
    Calculate Loss
         ↓
    Backpropagation
         ↓
    Calculate Gradients
         ↓
    Update Weights

Therefore:

    4 mini-batches
    4 weight updates
    1 epoch

Suppose the gradients from two samples in one mini-batch are:

    gradient 1 = 2
    gradient 2 = 6

The mini-batch gradient is their average:

    gradient = (2 + 6) / 2

    gradient = 4

Suppose:

    old weight = 5
    learning rate = 0.1

Weight update:

    new weight = old weight - learning_rate × gradient

    new weight = 5 - (0.1 × 4)

    new weight = 4.6

```python
import numpy as np

X = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)
y = np.array([2, 4, 6, 8, 10, 12, 14, 16], dtype=float)

w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 100
batch_size = 2

n = len(X)

for epoch in range(epochs):

    # Shuffle the dataset
    indices = np.random.permutation(n)

    X_shuffled = X[indices]
    y_shuffled = y[indices]

    # Process mini-batches
    for start in range(0, n, batch_size):

        end = start + batch_size

        X_batch = X_shuffled[start:end]
        y_batch = y_shuffled[start:end]

        # Forward pass
        y_pred = w * X_batch + b

        # Error
        error = y_pred - y_batch

        # Loss
        loss = np.mean(error ** 2)

        # Gradients
        dw = (2 / len(X_batch)) * np.sum(error * X_batch)
        db = (2 / len(X_batch)) * np.sum(error)

        # Update weights
        w = w - learning_rate * dw
        b = b - learning_rate * db

print("Weight:", w)
print("Bias:", b)
```
