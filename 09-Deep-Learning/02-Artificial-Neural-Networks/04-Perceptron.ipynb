# Perceptron

A Perceptron is one of the simplest types of artificial neural
networks and is considered one of the fundamental building blocks
of modern neural networks.

The Perceptron was introduced by Frank Rosenblatt in 1957.

It is mainly used for binary classification problems.

A Perceptron takes input features, applies weights and bias,
calculates a weighted sum, and then uses a step activation function
to produce a binary output.

## Perceptron Components

A Perceptron contains:

1. Input Features
2. Weights
3. Bias
4. Weighted Sum
5. Step Activation Function
6. Output

The weights and bias are learned during training.

## Step Activation Function

The original Perceptron uses a step activation function.

A simple step function can be defined as:

f(z) = 1, if z ≥ 0
f(z) = 0, if z < 0

Therefore:

If z ≥ 0 → Output = 1

If z < 0 → Output = 0

This allows the Perceptron to perform binary classification.

```python
x1 = 2
x2 = 3

w1 = 0.5
w2 = 0.8

b = -2

z = (w1 * x1) + (w2 * x2) + b

print("Weighted Sum:", z)

if z >= 0:
    output = 1
else:
    output = 0

print("Prediction:", output)
```

## Perceptron Decision Boundary

A Perceptron creates a decision boundary between two classes.

For two features:

w₁x₁ + w₂x₂ + b = 0

This equation represents the decision boundary.

The Perceptron predicts:

Class 1 → one side of the boundary

Class 0 → the other side of the boundary

## How Does a Perceptron Learn?

The Perceptron learns by updating its weights and bias when
it makes an incorrect prediction.

The general process is:

1. Initialize weights and bias.
2. Take an input sample.
3. Calculate the weighted sum.
4. Apply the step function.
5. Compare prediction with actual target.
6. Calculate the error.
7. Update weights and bias.
8. Repeat for the training data.

## Perceptron Error

The prediction error can be calculated as:

Error = Actual - Prediction

Mathematically:

e = y - ŷ

Where:

y = Actual target
ŷ = Predicted output
e = Error

## Perceptron Weight Update

The Perceptron updates its weights using:

wᵢ(new) = wᵢ(old) + η × error × xᵢ

The bias is updated using:

b(new) = b(old) + η × error

Where:

η = Learning Rate
error = Actual - Prediction
xᵢ = Input feature

## Perceptron vs Artificial Neuron

A Perceptron is a specific type of artificial neuron.

Basic artificial neuron:

Weighted Sum
      ↓
Activation Function
      ↓
Output

Perceptron:

Weighted Sum
      ↓
Step Function
      ↓
Binary Output

Modern neural networks use different activation functions
and much more flexible architectures.
