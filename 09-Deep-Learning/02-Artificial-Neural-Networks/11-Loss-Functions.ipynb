# Loss Functions

A Loss Function is a mathematical function that measures the
difference between the model's prediction and the actual target.

In simple words:

Loss Function = How wrong is the model?

During training, the neural network tries to minimize the loss.

General idea:

Actual Value
     ↓
Compare
     ↑
Predicted Value
     ↓
Loss

# Actual vs Predicted

Let:

y = Actual Value

ŷ = Predicted Value

The loss function calculates:

Loss = L(y, ŷ)

The goal of training is:

Minimize Loss

The neural network does this by updating its:

- Weights
- Biases

# Mean Squared Error (MSE)

MSE is commonly used for regression problems.

Formula:

MSE = (1/n) Σ(yᵢ - ŷᵢ)²

Where:

yᵢ = Actual value
ŷᵢ = Predicted value
n = Number of observations

The error is squared so that:

- Negative errors become positive.
- Large errors receive more penalty.

# MSE Example

Suppose:

Actual = [10, 20, 30]

Predicted = [12, 18, 33]

Errors:

10 - 12 = -2
20 - 18 = 2
30 - 33 = -3

Squared errors:

(-2)² = 4
(2)² = 4
(-3)² = 9

MSE:

MSE = (4 + 4 + 9) / 3

MSE = 17 / 3

MSE ≈ 5.67

```python
import numpy as np
from sklearn.metrics import mean_squared_error

y_true = np.array([10, 20, 30])
y_pred = np.array([12, 18, 33])

mse = mean_squared_error(y_true, y_pred)

print("MSE:", mse)
```

# Mean Absolute Error (MAE)

MAE calculates the average absolute difference between
actual and predicted values.

Formula:

MAE = (1/n) Σ|yᵢ - ŷᵢ|

Unlike MSE, MAE does not square the error.

Example:

Actual:

[10, 20, 30]

Predicted:

[12, 18, 33]

Errors:

[-2, 2, -3]

Absolute errors:

[2, 2, 3]

MAE:

MAE = (2 + 2 + 3) / 3

MAE = 7 / 3

MAE ≈ 2.33

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_true, y_pred)

print("MAE:", mae)
```

# Binary Cross-Entropy (BCE)

Binary Cross-Entropy is commonly used for binary classification.

Example:

Spam vs Not Spam

Classes:

0 = Not Spam
1 = Spam

The output layer commonly uses:

Sigmoid

which produces a probability between 0 and 1.

Formula:

L = -[y log(ŷ) + (1-y) log(1-ŷ)]

# Binary Cross-Entropy Example

Suppose:

Actual:

y = 1

Prediction:

ŷ = 0.9

The loss is:

L = -[1 × log(0.9) + (1-1) × log(1-0.9)]

L = -log(0.9)

L ≈ 0.105

This is a small loss because the model predicted a value
close to the actual class.

```python
import tensorflow as tf

y_true = tf.constant([1.0])
y_pred = tf.constant([0.9])

bce = tf.keras.losses.BinaryCrossentropy()

loss = bce(y_true, y_pred)

print("Binary Cross-Entropy:", loss.numpy())
```

# Categorical Cross-Entropy

Categorical Cross-Entropy is commonly used for multiclass
classification when the targets are one-hot encoded.

Example classes:

Cat
Dog
Horse

Actual:

[1, 0, 0]

Meaning:

Cat = 1
Dog = 0
Horse = 0

Suppose the model predicts:

[0.8, 0.1, 0.1]

The model gives the highest probability to Cat.

Categorical Cross-Entropy measures how well these probabilities
match the actual class.

# Formula

Categorical Cross-Entropy:

L = -Σ yᵢ log(ŷᵢ)

Where:

yᵢ = Actual class indicator
ŷᵢ = Predicted probability

For one-hot encoded targets, only the probability assigned to
the correct class directly contributes to the loss.

# Huber Loss

Huber Loss is commonly used for regression.

It combines properties of:

- MSE
- MAE

For small errors, it behaves like MSE.

For large errors, it behaves more like MAE.

This makes Huber Loss less sensitive to extreme outliers
than MSE.

# Loss vs Epoch

During training, we generally want the loss to decrease.

Example:

Epoch 1  → Loss = 1.50
Epoch 2  → Loss = 1.10
Epoch 3  → Loss = 0.75
Epoch 4  → Loss = 0.50
Epoch 5  → Loss = 0.32

The decreasing loss indicates that the model is learning
to reduce its training error.

However, training loss alone is not enough to determine whether
the model generalizes well.
