# Loss Function

A loss function measures how wrong the neural network's
prediction is compared with the actual target.

The model makes a prediction:

Input
  ↓
Neural Network
  ↓
Prediction

Then the loss function compares:

Prediction ↔ Actual Value

A smaller loss generally means the predictions are closer
to the target values.

During training, the goal is to minimize the loss.

## Simple Loss Example

Suppose the actual value is:

y = 1

The model predicts:

Prediction = 0.9

The prediction is close to the actual value,
so the loss should be relatively small.

If the model predicts:

Prediction = 0.1

The prediction is far from the actual value,
so the loss should be larger.

## Common PyTorch Loss Functions

For different problems, we use different loss functions.

### Regression

Mean Squared Error:

nn.MSELoss()

### Binary Classification

Binary Cross Entropy:

nn.BCELoss()

### Binary Classification with logits

nn.BCEWithLogitsLoss()

### Multi-class Classification

Cross Entropy:

nn.CrossEntropyLoss()

## Mean Squared Error

MSE is commonly used for regression.

It calculates the average squared difference
between actual and predicted values.

The basic idea is:

Error = Actual - Prediction

Then:

Squared Error = (Actual - Prediction)²

Finally, average the squared errors.

Large errors receive a larger penalty because
the error is squared.

```python
import torch
import torch.nn as nn

actual = torch.tensor([10.0, 20.0, 30.0])
predicted = torch.tensor([12.0, 18.0, 29.0])

loss_function = nn.MSELoss()

loss = loss_function(predicted, actual)

print("MSE Loss:", loss.item())
```

## Binary Cross Entropy

Binary Cross Entropy is used for binary classification.

The target contains two classes:

0
1

For example:

Spam     → 1
Not Spam → 0

The neural network usually outputs a probability
between 0 and 1.

Example:

0.95 → strong probability of class 1
0.10 → strong probability of class 0

Binary Cross Entropy measures how well these
probabilities match the actual labels.

```python
actual = torch.tensor([
    [1.0],
    [0.0],
    [1.0]
])

predicted = torch.tensor([
    [0.9],
    [0.1],
    [0.8]
])

loss_function = nn.BCELoss()

loss = loss_function(predicted, actual)

print("BCE Loss:", loss.item())
```

# Optimizer

The optimizer updates the model's trainable parameters
(weights and biases) to reduce the loss.

The basic process is:

Prediction
    ↓
Loss
    ↓
Gradients
    ↓
Optimizer
    ↓
Updated Weights

## Gradient Descent

Suppose a model has a weight:

w

After backpropagation, we calculate:

gradient

Gradient tells us how the loss changes
when the weight changes.

The basic update rule is:

new weight =
old weight - learning rate × gradient

The learning rate controls how large
the update should be.

## Learning Rate

The learning rate controls the size of weight updates.

Small learning rate:

- Training can be slow
- More updates may be required

Large learning rate:

- Training can become unstable
- The model may overshoot good solutions

A common starting value for Adam is:

0.001

```python
learning_rate = 0.001

print(learning_rate)
```

## Adam Optimizer

Adam stands for Adaptive Moment Estimation.

Adam is one of the most commonly used optimizers
in deep learning.

It adapts the parameter updates using information
from previous gradients.

PyTorch provides:

torch.optim.Adam()

## SGD

SGD stands for Stochastic Gradient Descent.

It is a basic and important optimization algorithm.

PyTorch:

torch.optim.SGD()
