# PyTorch Introduction

PyTorch is an open-source deep learning framework developed by Meta AI.

It is used to build, train, evaluate, and deploy machine learning
and deep learning models.

PyTorch provides tools for:

- Tensor operations
- Automatic differentiation
- Neural networks
- Backpropagation
- GPU acceleration
- Model training
- Computer Vision
- NLP
- Generative AI

## PyTorch vs TensorFlow

Both PyTorch and TensorFlow are popular deep learning frameworks.

### PyTorch

- Developed by Meta AI
- Python-friendly
- Uses tensors
- Supports automatic differentiation
- Uses `torch.nn` for neural networks
- Uses `torch.optim` for optimizers
- Very popular in research and modern AI

### TensorFlow

- Developed by Google
- Uses tensors
- Supports automatic differentiation
- Uses Keras for high-level neural networks
- Strong production and deployment ecosystem

Both frameworks can be used to build and train neural networks.
