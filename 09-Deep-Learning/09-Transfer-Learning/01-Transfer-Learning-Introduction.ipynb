# Transfer Learning — Introduction

Transfer Learning is a Deep Learning technique where we take a model that has already been trained on a large dataset and reuse its learned knowledge for a new task.

Instead of training a neural network completely from scratch, we start with a pretrained model and adapt it to our own dataset.

## What is Transfer Learning?

Transfer Learning means:

> Take knowledge learned from one problem and reuse it to solve another related problem.

For example, suppose a CNN has been trained on millions of images.

During training, the CNN learns useful visual features such as:

- Edges
- Lines
- Corners
- Textures
- Shapes
- Patterns
- Object parts

We can reuse these learned features for our own image classification problem.

For example:

A pretrained CNN → trained on a large image dataset
## Why Do We Need Transfer Learning?

Training a deep CNN from scratch can require:

1. A large dataset
2. Powerful GPU hardware
3. A lot of training time
4. Many computational resources
5. Careful hyperparameter tuning

If we only have a few thousand images, training a large CNN from scratch may not be the best choice.

Transfer Learning allows us to start with a model that has already learned useful features.

## What Does a Pretrained CNN Learn?

A CNN learns different types of features at different layers.

### Early layers

Usually learn simple features:

- Edges
- Lines
- Corners
- Basic textures

### Middle layers

Learn more complex patterns:

- Shapes
- Textures
- Object parts

### Deeper layers

Learn more task-specific features:

- Faces
- Eyes
- Ears
- Wheels
- Animal parts
- Complex object patterns

This is one of the main reasons Transfer Learning works.

The early and middle features learned by a CNN can often be useful for many different image tasks.

## Transfer Learning Pipeline

The basic Transfer Learning workflow is:

1. Select a pretrained model
2. Load its pretrained weights
3. Remove or ignore its original classifier
4. Freeze the pretrained layers
5. Add our own classification layers
6. Train the new classifier
7. Evaluate the model
8. Optionally unfreeze some pretrained layers
9. Fine-tune the model with a small learning rate

The two major stages are:

### Stage 1 — Feature Extraction

Use the pretrained CNN as a fixed feature extractor.

### Stage 2 — Fine-Tuning

Unfreeze some pretrained layers and train them together with the new classifier.

## Feature Extraction

Feature Extraction means using the pretrained CNN to extract useful features while keeping its pretrained weights frozen.

Example:

Pretrained CNN
      ↓
Frozen layers
      ↓
Extract features
      ↓
New classifier
      ↓
Prediction

During this stage, the pretrained CNN's weights do NOT change.

Only the newly added classification layers are trained.

This prevents the pretrained model from changing too aggressively and destroying useful previously learned features.
