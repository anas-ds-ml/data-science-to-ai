# Input, Hidden and Output Layers

An Artificial Neural Network (ANN) is organized into different
layers.

The three main types of layers are:

1. Input Layer
2. Hidden Layer
3. Output Layer

Basic flow:

Input Layer
     ↓
Hidden Layer(s)
     ↓
Output Layer

# 1. Input Layer

The Input Layer is the first layer of a neural network.

It receives the features from the dataset.

For example, suppose we want to predict house prices.

Our features are:

- Area
- Number of Bedrooms
- Number of Bathrooms
- Age of House

Therefore:

Number of Input Features = 4

The network will receive:

x₁ = Area
x₂ = Bedrooms
x₃ = Bathrooms
x₄ = Age

## Important Point About Input Layer

The number of input neurons is generally determined by the number
of input features.

If our dataset contains:

10 features

then the input layer receives:

10 input values

Example:

X.shape = (1000, 10)

This means:

1000 → Number of samples
10   → Number of features

Therefore, the input dimension is:

10

# 2. Hidden Layer

A Hidden Layer is a layer between the input layer and output layer.

Example:

Input Layer
     ↓
Hidden Layer
     ↓
Output Layer

A neural network can have:

- One hidden layer
- Multiple hidden layers

Each hidden layer contains neurons.

These neurons learn patterns and relationships from the input data.

## Why is it Called "Hidden"?

It is called a hidden layer because its internal computations
are not directly provided as input or final output.

The network receives:

Input

and produces:

Output

The transformations between them happen inside hidden layers.

Example:

Input
  ↓
Hidden Layer 1
  ↓
Hidden Layer 2
  ↓
Output

# What Does a Hidden Neuron Do?

Each neuron in a hidden layer performs:

1. Weighted Sum
2. Add Bias
3. Apply Activation Function

Mathematically:

z = Σ(wᵢxᵢ) + b

Then:

a = f(z)

Where:

x = Inputs
w = Weights
b = Bias
f = Activation Function
a = Neuron Output

# Multiple Hidden Layers

A neural network can contain several hidden layers.

Example:

Input Layer
     ↓
Hidden Layer 1
     ↓
Hidden Layer 2
     ↓
Hidden Layer 3
     ↓
Output Layer

When a network contains many layers of transformations,
it is commonly called a Deep Neural Network (DNN).

# 3. Output Layer

The Output Layer is the final layer of the neural network.

It produces the final prediction.

The structure of the output layer depends on the type of problem.

The three important cases are:

1. Regression
2. Binary Classification
3. Multiclass Classification

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),

    tf.keras.layers.Dense(16, activation="relu"),

    tf.keras.layers.Dense(8, activation="relu"),

    tf.keras.layers.Dense(1, activation="sigmoid")
])

model.summary()
```
