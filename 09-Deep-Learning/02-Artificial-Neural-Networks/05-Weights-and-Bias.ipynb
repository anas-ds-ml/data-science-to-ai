# Weights and Bias in Artificial Neural Networks

Weights and biases are two of the most important learnable
parameters in an Artificial Neural Network (ANN).

A neuron uses weights and bias to transform input data into
an output.

The basic equation is:

z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b

Where:

- x = Input
- w = Weight
- b = Bias
- z = Weighted Sum

## What is a Weight?

A weight is a learnable parameter that determines how strongly
an input influences a neuron.

Every input generally has a corresponding weight.

Example:

x₁ → w₁
x₂ → w₂
x₃ → w₃

The weighted contributions are:

x₁ × w₁
x₂ × w₂
x₃ × w₃

The neural network learns the appropriate weight values during
training.

## Intuition Behind Weights

Think of weights as the importance given to different features.

Suppose we want to predict whether a student will pass.

Features:

- Study Hours
- Attendance
- Previous Marks

The model might learn:

Study Hours    → Large positive weight
Attendance     → Medium positive weight
Previous Marks → Large positive weight

A larger positive weight means that the feature has a stronger
positive influence on the neuron's output.

A negative weight means that the feature can have a negative
influence on the output.

## Positive and Negative Weights

Weights can be:

- Positive
- Negative
- Close to zero

### Positive Weight

A positive weight increases the weighted sum when the input
increases.

### Negative Weight

A negative weight decreases the weighted sum when the input
increases.

### Weight Near Zero

A weight close to zero means that the corresponding input has
relatively little influence on that neuron.

```python
x1 = 5
x2 = 2

w1 = 0.8
w2 = -0.5

weighted_sum = (x1 * w1) + (x2 * w2)

print("Weighted Sum:", weighted_sum)
```

## What is Bias?

Bias is an additional learnable parameter added to the weighted
sum.

The equation becomes:

z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b

Bias allows the neuron to shift its output independently of the
input values.

Without bias:

z = Σ(wᵢxᵢ)

With bias:

z = Σ(wᵢxᵢ) + b

## Intuition Behind Bias

A useful way to think about bias is as an adjustable offset.

Consider a simple equation:

y = wx + b

Here:

w controls the slope.

b controls the intercept.

Therefore, bias allows the model to shift the decision boundary
or activation function.

This gives the neuron additional flexibility.

```python
x1 = 2
x2 = 3

w1 = 0.5
w2 = 0.8

b = 1

z = (w1 * x1) + (w2 * x2) + b

print("Weighted Sum + Bias:", z)
```
