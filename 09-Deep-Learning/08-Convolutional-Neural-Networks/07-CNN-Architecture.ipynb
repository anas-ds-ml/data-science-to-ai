# CNN Architecture

A Convolutional Neural Network (CNN) is usually built using several
different types of layers.

A basic CNN architecture is:

Input Image
    ↓
Convolution
    ↓
Activation Function (ReLU)
    ↓
Pooling
    ↓
Convolution
    ↓
Activation Function (ReLU)
    ↓
Pooling
    ↓
Flatten
    ↓
Fully Connected (Dense) Layers
    ↓
Output Layer

Each layer has a specific job.

- Convolution → extracts features
- ReLU → introduces non-linearity
- Pooling → reduces spatial dimensions
- Flatten → converts feature maps into a vector
- Dense → performs classification
- Output → produces the final prediction

## 1. Basic CNN Architecture

Let's consider an RGB image of:

128 × 128 × 3

The CNN can process it like this:

Input
128 × 128 × 3
        ↓
Conv2D
128 × 128 × 32
        ↓
ReLU
        ↓
MaxPooling
64 × 64 × 32
        ↓
Conv2D
64 × 64 × 64
        ↓
ReLU
        ↓
MaxPooling
32 × 32 × 64
        ↓
Flatten
65,536 values
        ↓
Dense
128 neurons
        ↓
Output
Number of classes

As we go deeper into the network:

- Height decreases
- Width decreases
- Number of feature maps usually increases
- Features become more complex

## 2. Input Layer

The input layer receives the image.

For an RGB image:

128 × 128 × 3

Where:

- 128 = image height
- 128 = image width
- 3 = RGB channels

The CNN expects a batch dimension as well.

For example:

(32, 128, 128, 3)

Where:

- 32 = batch size
- 128 = height
- 128 = width
- 3 = channels

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(128, 128, 3))
])

model.summary()
```

## 3. Convolution Layer

The convolution layer extracts useful features from an image.

Early convolution layers may learn:

- Edges
- Lines
- Corners
- Simple textures

Example:

Conv2D(
    filters=32,
    kernel_size=(3,3)
)

This means the layer learns 32 different filters.

Each filter can learn to detect a different pattern.

```python
model = tf.keras.Sequential([
    
    tf.keras.layers.Input(shape=(128, 128, 3)),
    
    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    )
])

model.summary()
```

## 4. ReLU Activation

ReLU stands for:

Rectified Linear Unit

The function is:

ReLU(x) = max(0, x)

It converts negative values into zero while keeping positive values.

ReLU introduces non-linearity into the CNN.

In Keras we commonly use:

activation="relu"

```python
import numpy as np

x = np.array([-5, -2, 0, 2, 5])

relu_output = np.maximum(0, x)

print("Input :", x)
print("ReLU  :", relu_output)
```

## 5. Pooling Layer

Pooling reduces the spatial dimensions of feature maps.

The most common pooling layer is:

MaxPooling2D

A typical configuration is:

pool_size = (2,2)

For example:

128 × 128 × 32
        ↓
MaxPooling
        ↓
64 × 64 × 32

Notice:

Height: 128 → 64
Width: 128 → 64
Channels: 32 → 32

Pooling reduces height and width but normally keeps
the number of channels unchanged.

```python
model = tf.keras.Sequential([
    
    tf.keras.layers.Input(shape=(128, 128, 3)),
    
    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    )
])

model.summary()
```

## 6. Building Deeper Convolution Blocks

CNNs usually contain multiple convolution blocks.

Example:

Block 1:

128 × 128 × 3
      ↓
Conv2D 32
      ↓
MaxPooling
      ↓
64 × 64 × 32

Block 2:

64 × 64 × 32
      ↓
Conv2D 64
      ↓
MaxPooling
      ↓
32 × 32 × 64

Notice that the number of filters increased:

32 → 64

At deeper levels, the network can learn more complex features.

```python
model = tf.keras.Sequential([
    
    tf.keras.layers.Input(shape=(128, 128, 3)),
    
    # Convolution Block 1
    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),
    
    # Convolution Block 2
    tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    )
])

model.summary()
```

## 7. Feature Hierarchy

One of the most important ideas in CNNs is hierarchical feature
learning.

### Early Layers

Learn simple features:

- Edges
- Lines
- Corners

### Middle Layers

Combine simple features into:

- Textures
- Curves
- Shapes
- Patterns

### Deep Layers

Learn more complex structures:

- Eyes
- Wheels
- Faces
- Object parts

### Final Layers

Use these learned features to classify the object.

Example:

Image of a cat

Edges
  ↓
Curves
  ↓
Eyes / ears / fur
  ↓
Cat features
  ↓
Cat prediction

## 8. Flatten Layer

After convolution and pooling, we have 3D feature maps.

For example:

32 × 32 × 64

Dense layers expect a one-dimensional vector.

Flatten converts:

32 × 32 × 64

into:

65,536

because:

32 × 32 × 64 = 65,536

Flatten does not learn features.
It simply rearranges the values into a 1D vector.

```python
model = tf.keras.Sequential([
    
    tf.keras.layers.Input(shape=(128, 128, 3)),
    
    tf.keras.layers.Conv2D(
        32, (3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(2, 2),
    
    tf.keras.layers.Conv2D(
        64, (3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(2, 2),
    
    tf.keras.layers.Flatten()
])

model.summary()
```

## 9. Fully Connected / Dense Layer

After Flatten, the feature vector is passed to Dense layers.

Dense layers combine the features learned by the convolution layers.

Example:

Flatten
65,536 values
      ↓
Dense
128 neurons
      ↓
Output

A Dense layer can learn relationships between the extracted features
and the target classes.

```python
model = tf.keras.Sequential([
    
    tf.keras.layers.Input(shape=(128, 128, 3)),
    
    tf.keras.layers.Conv2D(
        32, (3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(2, 2),
    
    tf.keras.layers.Conv2D(
        64, (3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(2, 2),
    
    tf.keras.layers.Flatten(),
    
    tf.keras.layers.Dense(
        128,
        activation="relu"
    )
])

model.summary()
```

## 10. Output Layer

The output layer depends on the type of classification problem.

### Binary Classification

Two classes:

Cat vs Dog

Use:

Dense(1, activation="sigmoid")

### Multi-Class Classification

Example:

Cat
Dog
Horse
Bird

There are 4 classes.

Use:

Dense(4, activation="softmax")

The number of output neurons should equal the number of classes.

### Regression

For regression:

Dense(1)

Usually no activation or linear activation is used.

```python
binary_output = tf.keras.layers.Dense(
    1,
    activation="sigmoid"
)

print(binary_output)
```

```python
number_of_classes = 10

multiclass_output = tf.keras.layers.Dense(
    number_of_classes,
    activation="softmax"
)

print(multiclass_output)
```

```python
model = tf.keras.Sequential([
    
    # Input
    tf.keras.layers.Input(shape=(128, 128, 3)),
    
    # Convolution Block 1
    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),
    
    # Convolution Block 2
    tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),
    
    # Convolution Block 3
    tf.keras.layers.Conv2D(
        filters=128,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),
    
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),
    
    # Classifier
    tf.keras.layers.Flatten(),
    
    tf.keras.layers.Dense(
        128,
        activation="relu"
    ),
    
    tf.keras.layers.Dropout(0.5),
    
    # Binary Classification
    tf.keras.layers.Dense(
        1,
        activation="sigmoid"
    )
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```
