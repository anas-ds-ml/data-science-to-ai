# Artificial Neuron

An Artificial Neuron is the basic computational unit of an
Artificial Neural Network (ANN).

It receives one or more input values, multiplies each input by
a corresponding weight, adds a bias, and passes the result
through an activation function.

The basic structure is:

Inputs
   ↓
Weights
   ↓
Weighted Sum
   ↓
Add Bias
   ↓
Activation Function
   ↓
Output

## Basic Mathematical Representation

An artificial neuron can be represented mathematically as:

z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b

After calculating the weighted sum, an activation function is applied:

a = f(z)

Where:

- x = Input
- w = Weight
- b = Bias
- z = Weighted Sum
- f = Activation Function
- a = Final Output

## Components of an Artificial Neuron

An artificial neuron contains the following important components:

1. Inputs
2. Weights
3. Weighted Sum
4. Bias
5. Activation Function
6. Output

These components work together to transform input data into
an output value.

## 1. Inputs

Inputs are the features provided to the artificial neuron.

For example, suppose we want to predict whether a student
will pass an exam.

We may have:

x₁ = Study Hours
x₂ = Attendance
x₃ = Previous Marks

These values are provided to the neuron as inputs.

## 2. Weights

Each input has a corresponding weight.

Weights determine how strongly each input influences the neuron.

For three inputs:

x₁ → w₁
x₂ → w₂
x₃ → w₃

The weighted contribution of each input is:

x₁w₁
x₂w₂
x₃w₃

During training, the neural network learns appropriate
weight values.

## 3. Weighted Sum

The neuron multiplies each input by its corresponding weight
and then adds all the values together.

The equation is:

z = w₁x₁ + w₂x₂ + w₃x₃

For n inputs:

z = Σ(wᵢxᵢ)

This is called the weighted sum.

## 4. Bias

Bias is an additional learnable parameter added to the
weighted sum.

The complete equation becomes:

z = Σ(wᵢxᵢ) + b

Bias allows the neuron to shift its activation and helps the
network learn more flexible relationships.

The bias is also learned during training.

## 5. Activation Function

After calculating the weighted sum and adding the bias,
the result is passed through an activation function.

a = f(z)

The activation function introduces non-linearity into the neural
network.

Common activation functions include:

- ReLU
- Sigmoid
- Tanh
- Softmax

## Complete Artificial Neuron

The complete mathematical process is:

Inputs
   ↓
Multiply each input by its weight
   ↓
Calculate weighted sum
   ↓
Add bias
   ↓
Apply activation function
   ↓
Output

Mathematically:

z = Σ(wᵢxᵢ) + b

a = f(z)
