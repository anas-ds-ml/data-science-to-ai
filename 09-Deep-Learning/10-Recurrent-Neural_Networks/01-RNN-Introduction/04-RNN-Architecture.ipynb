# RNN Architecture

## What is RNN Architecture?

A Recurrent Neural Network (RNN) is a neural network designed to work with **sequential data**.

Unlike a traditional feed-forward neural network, an RNN keeps information from the previous time step and uses it while processing the current input.

The basic idea is:

**Current Input + Previous Hidden State → RNN Cell → New Hidden State**

The new hidden state is then passed to the next time step.

### Basic Flow

```text
Previous Hidden State
        hₜ₋₁
          │
          ▼
Current Input ──► [ RNN Cell ] ──► New Hidden State
        xₜ             │                 hₜ
                       │
                       ▼
                    Output
                       yₜ

# Unrolled RNN Architecture

To understand how an RNN processes a sequence, we can **unroll** the RNN across multiple time steps.

Suppose the input sequence contains four inputs:

```text
x₁, x₂, x₃, x₄

# Complete RNN Architecture

The complete basic RNN architecture can be represented as:

```text
                    Previous Hidden State
                           hₜ₋₁
                             │
                             │
                             ▼
Current Input xₜ ───────► [ RNN CELL ]
                             │
                             │
                             ▼
                    New Hidden State
                           hₜ
                             │
                             ▼
                       [ Output Layer ]
                             │
                             ▼
                          Output
                           yₜ

````python
import numpy as np

# Sequence of 4 simple input values
x = [1.0, 2.0, 3.0, 4.0]

# Simple RNN parameters
W_xh = 0.5
W_hh = 0.8
b_h = 0.1

# Initial hidden state
h = 0.0

print("RNN Hidden State Flow:\n")

for t, x_t in enumerate(x, start=1):

    # RNN hidden-state equation
    h = np.tanh(W_xh * x_t + W_hh * h + b_h)

    print(f"Time Step {t}")
    print(f"Input (x{t})        : {x_t}")
    print(f"Hidden State (h{t}) : {h:.4f}")
    print("-" * 30)
````
