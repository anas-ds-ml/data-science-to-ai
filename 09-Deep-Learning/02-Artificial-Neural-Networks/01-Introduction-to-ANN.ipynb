# Introduction to Artificial Neural Networks (ANN)

Artificial Neural Network (ANN) is a Machine Learning model inspired
by the structure and functioning of biological neural networks.

ANNs consist of interconnected artificial neurons that work together
to learn patterns from data and make predictions.

Artificial Neural Networks are one of the fundamental building blocks
of Deep Learning.

## What is an Artificial Neural Network?

An Artificial Neural Network is a computational model made up of
interconnected nodes called neurons.

These neurons are organized into layers:

- Input Layer
- Hidden Layer(s)
- Output Layer

The neurons receive input data, perform mathematical operations,
and pass information to the next layer.

A basic ANN looks like:

Input Layer
     ↓
Hidden Layer
     ↓
Output Layer

## 1. Input Layer

The input layer receives the features from the dataset.

For example, suppose we want to predict whether a customer
will purchase a product.

Our features might be:

- Age
- Income
- Previous Purchases

Then the input layer may contain three input values:

X₁ = Age
X₂ = Income
X₃ = Previous Purchases

The input layer passes these values to the hidden layers.

## 2. Hidden Layers

Hidden layers are the intermediate layers between the input
and output layers.

They perform mathematical transformations on the input data.

An ANN can have:

- One hidden layer
- Multiple hidden layers
- Many hidden layers

When a neural network contains multiple hidden layers,
it is commonly referred to as a Deep Neural Network (DNN).

Hidden layers allow the network to learn increasingly complex
patterns from the input data.

## 3. Output Layer

The output layer produces the final prediction of the network.

The structure of the output layer depends on the problem.

### Binary Classification

Example:

Spam or Not Spam

Output:

0 → Not Spam
1 → Spam

### Multiclass Classification

Example:

Cat, Dog, Horse

The output layer can contain one output for each class.

### Regression

Example:

House Price Prediction

The output layer can contain a single numerical output.

## Artificial Neuron

The basic computational unit of an ANN is called an artificial neuron.

A neuron receives one or more inputs.

Each input is multiplied by a corresponding weight.
The weighted inputs are then added together with a bias.

Mathematically:

z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b

The result is then passed through an activation function:

a = f(z)

Where:

- x = Input
- w = Weight
- b = Bias
- z = Weighted Sum
- f = Activation Function
- a = Neuron Output

## Weights

Weights determine how strongly each input influences a neuron.

For example:

z = w₁x₁ + w₂x₂

If w₁ is large, x₁ has a stronger influence on the neuron.

During training, the neural network learns appropriate weight
values from the data.

## Bias

Bias is an additional learnable parameter added to the weighted sum.

The basic equation is:

z = wx + b

Bias allows the neuron to shift its activation and helps the network
learn more flexible relationships.

Weights and biases are learned during the training process.

## Activation Function

An activation function determines the output of a neuron after
the weighted sum and bias have been calculated.

Without activation functions, a neural network would be limited
to learning linear relationships.

Activation functions introduce non-linearity into the network.

Common activation functions include:

- ReLU
- Sigmoid
- Tanh
- Softmax

## How ANN Learns

An ANN learns by adjusting its weights and biases during training.

The general learning process is:

Input Data
    ↓
Weighted Sum
    ↓
Activation Function
    ↓
Prediction
    ↓
Loss Calculation
    ↓
Backpropagation
    ↓
Gradient Calculation
    ↓
Weight Update
    ↓
Improved Prediction

This process is repeated many times during training.
