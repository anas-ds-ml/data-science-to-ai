# PyTorch Autograd

Autograd stands for Automatic Differentiation.

PyTorch Autograd automatically calculates gradients
of tensors.

Gradients are required for:

- Backpropagation
- Gradient Descent
- Weight Updates
- Training Neural Networks

The basic flow is:

Forward Pass
     ↓
Calculate Loss
     ↓
Backward Pass
     ↓
Calculate Gradients
     ↓
Update Weights

## Why Do We Need Gradients?

During ANN training, the model has weights and biases.

Initially, these parameters are usually not perfect.

The model makes a prediction:

Prediction
    ↓
Compare with Actual Value
    ↓
Loss

We then need to know:

"How should each weight change to reduce the loss?"

The gradient tells us the direction and rate of change.

A simplified weight update is:

new_weight = old_weight - learning_rate × gradient

## requires_grad=True

If we want PyTorch to calculate the gradient
of a tensor, we can set:

requires_grad=True

This tells PyTorch:

"Track the operations performed on this tensor
because I may need its gradient later."

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

print(x)
print(x.requires_grad)
```

## Calculating a Gradient

Suppose:

y = x²

The derivative is:

dy/dx = 2x

If:

x = 2

Then:

dy/dx = 4

PyTorch can calculate this automatically.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2

print("y =", y)
```

```python
y.backward()

print("Gradient:", x.grad)
```

## What does backward() do?

The backward() function calculates gradients.

When we write:

y.backward()

PyTorch goes backward through the computation graph
and calculates how much y changes with respect to
the tensors that have requires_grad=True.

For neural networks, this is the basis of backpropagation.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = 3 * x**2 + 2 * x

y.backward()

print("y:", y)
print("dy/dx:", x.grad)
```

## Multiple Variables

Suppose:

z = x² + y²

We want:

dz/dx
dz/dy

```python
import torch

x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)

z = x**2 + y**2

z.backward()

print("dz/dx:", x.grad)
print("dz/dy:", y.grad)
```

```python
import torch

# Input
x = torch.tensor(2.0)

# Weight
w = torch.tensor(0.5, requires_grad=True)

# Bias
b = torch.tensor(1.0, requires_grad=True)

# Forward pass
z = x * w + b

print("Output:", z)
```

```python
loss = z**2

loss.backward()

print("Weight gradient:", w.grad)
print("Bias gradient:", b.grad)
```

## Weight Updates

Suppose:

weight = 0.5

gradient = 8

learning_rate = 0.01

Gradient descent updates the weight:

new_weight = weight - learning_rate × gradient

Therefore:

new_weight = 0.5 - (0.01 × 8)

new_weight = 0.42

```python
learning_rate = 0.01

with torch.no_grad():
    w -= learning_rate * w.grad

print("Updated weight:", w)
```
