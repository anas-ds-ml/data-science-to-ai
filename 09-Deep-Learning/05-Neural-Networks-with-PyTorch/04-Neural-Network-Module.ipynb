# PyTorch Neural Network Module

PyTorch provides the `torch.nn` module for building neural networks.

The most important class for creating custom neural networks is:

    nn.Module

`nn.Module` is the base class for neural network models in PyTorch.

We create our own neural network by inheriting from `nn.Module`.

## Importing torch.nn

PyTorch's neural network functionality is available through:

    torch.nn

We commonly write:

    import torch.nn as nn

This allows us to use:

- nn.Module
- nn.Linear
- nn.ReLU
- nn.Sigmoid
- nn.Dropout
- nn.MSELoss
- nn.CrossEntropyLoss

```python
import torch
import torch.nn as nn

print(nn)
```

## What is nn.Module?

`nn.Module` is the base class for neural network models.

When we create a custom neural network, we inherit from it.

Basic structure:

class MyModel(nn.Module):

    def __init__(self):
        super().__init__()

    def forward(self, x):
        ...
        
The `__init__()` method defines the layers.

The `forward()` method defines how data moves
through those layers.

```python
import torch
import torch.nn as nn

class MyNeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(10, 64)
        self.layer2 = nn.Linear(64, 32)
        self.output = nn.Linear(32, 1)

    def forward(self, x):

        x = self.layer1(x)
        x = self.layer2(x)
        x = self.output(x)

        return x


model = MyNeuralNetwork()

print(model)
```

## The __init__() Method

The `__init__()` method is where we define
the components of our neural network.

For example:

self.layer1 = nn.Linear(10, 64)

means:

10 input features
        ↓
64 neurons

Similarly:

self.layer2 = nn.Linear(64, 32)

means:

64 neurons
        ↓
32 neurons

And:

self.output = nn.Linear(32, 1)

means:

32 neurons
        ↓
1 output neuron

## super().__init__()

When creating a custom neural network, we write:

super().__init__()

This initializes the parent `nn.Module` class.

It allows PyTorch to properly track:

- Layers
- Parameters
- Weights
- Biases
- Model state

It should be called inside `__init__()`.

## The forward() Method

The `forward()` method defines the forward propagation
of the neural network.

For example:

x = self.layer1(x)
x = self.layer2(x)
x = self.output(x)

Data flows:

Input
  ↓
Layer 1
  ↓
Layer 2
  ↓
Output

```python
def forward(self, x):

    x = self.layer1(x)
    x = self.layer2(x)
    x = self.output(x)

    return x
```

## Adding Activation Functions

Neural networks normally use activation functions
between layers.

We can use:

    nn.ReLU()

For binary classification, the final layer can use:

    nn.Sigmoid()

```python
import torch
import torch.nn as nn

class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(10, 64)
        self.layer2 = nn.Linear(64, 32)
        self.output = nn.Linear(32, 1)

        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        x = self.layer1(x)
        x = self.relu(x)

        x = self.layer2(x)
        x = self.relu(x)

        x = self.output(x)
        x = self.sigmoid(x)

        return x


model = NeuralNetwork()

print(model)
```

## Forward Pass

After creating the model, we can pass data through it.

Suppose we have:

100 samples
10 features

The input tensor shape is:

(100, 10)

After the first layer:

(100, 64)

After the second layer:

(100, 32)

Final output:

(100, 1)

```python
X = torch.randn(100, 10)

output = model(X)

print("Input shape:", X.shape)
print("Output shape:", output.shape)
```

## Calling the Model

We normally write:

    output = model(X)

instead of:

    output = model.forward(X)

PyTorch internally calls the `forward()` method
when we use:

    model(X)

This is the recommended way to use a PyTorch model.

## Model Parameters

Neural networks contain trainable parameters.

These include:

- Weights
- Biases

PyTorch automatically tracks parameters
defined inside `nn.Module`.

We can see them using:

    model.parameters()

```python
for name, parameter in model.named_parameters():

    print(name)
    print(parameter.shape)
    print()
```

## Counting Parameters

We can calculate the total number of trainable
parameters in the model.

For a Linear layer:

Parameters = input_features × output_features + output_features

The second term represents the biases.

```python
total_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

print("Total parameters:", total_parameters)
```

## nn.Sequential

If our network is a simple sequence of layers,
we can use `nn.Sequential`.

Instead of writing a custom `forward()` method,
PyTorch can automatically pass data through the layers.

```python
model = nn.Sequential(
    nn.Linear(10, 64),
    nn.ReLU(),

    nn.Linear(64, 32),
    nn.ReLU(),

    nn.Linear(32, 1),
    nn.Sigmoid()
)

print(model)
```

## Custom nn.Module vs nn.Sequential

### nn.Sequential

Useful for simple architectures.

Example:

Linear → ReLU → Linear → ReLU → Output

### Custom nn.Module

Useful for:

- Complex architectures
- Multiple branches
- Skip connections
- Custom forward logic
- Advanced models

For serious PyTorch projects, understanding
custom `nn.Module` is very important.

```python
import torch
import torch.nn as nn

class ANN(nn.Module):

    def __init__(self):
        super().__init__()

        self.fc1 = nn.Linear(10, 64)
        self.fc2 = nn.Linear(64, 32)
        self.fc3 = nn.Linear(32, 1)

        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        x = self.fc1(x)
        x = self.relu(x)

        x = self.fc2(x)
        x = self.relu(x)

        x = self.fc3(x)
        x = self.sigmoid(x)

        return x


model = ANN()

print(model)
```
