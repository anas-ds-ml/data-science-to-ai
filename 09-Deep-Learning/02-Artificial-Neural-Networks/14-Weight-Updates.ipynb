# Weight and Bias Updates

Neural networks learn by continuously updating their weights
and biases during training.

Initially, weights and biases are usually initialized with
some starting values.

During training:

Forward Propagation
        ↓
Prediction
        ↓
Loss Calculation
        ↓
Backpropagation
        ↓
Gradients
        ↓
Weight & Bias Updates

The goal is to update the parameters so that the loss decreases.

# What Are Parameters?

The trainable parameters of a neural network are mainly:

- Weights
- Biases

Weights determine the importance of input features.

Bias allows a neuron to shift its activation.

For a neuron:

z = wx + b

Where:

w = Weight
x = Input
b = Bias
z = Weighted Sum

# Weight Update Formula

The basic Gradient Descent update rule is:

w_new = w_old - η(∂L/∂w)

Where:

w_old = Current weight
w_new = Updated weight
η = Learning rate
∂L/∂w = Gradient of loss with respect to weight

The gradient tells us the direction and magnitude of the
local change in loss.

# Bias Update Formula

The bias is updated in the same way:

b_new = b_old - η(∂L/∂b)

Where:

b_old = Current bias
b_new = Updated bias
η = Learning rate
∂L/∂b = Gradient of loss with respect to bias

Therefore:

Weights → Updated using their gradients

Biases → Updated using their gradients
