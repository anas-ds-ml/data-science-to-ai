# Adam Optimization

Adam stands for:

    Adaptive Moment Estimation

Adam is one of the most widely used optimization algorithms
for training neural networks and deep learning models.

Adam combines two important ideas:

    1. Momentum
    2. RMSProp

Momentum helps remember the direction of previous gradients.

RMSProp helps adapt the learning rate based on the
magnitude of recent gradients.

Adam combines both ideas.

## Why Do We Need Adam?

We have already learned:

    Gradient Descent
    SGD
    Mini-Batch Gradient Descent
    Momentum
    RMSProp

Each improves optimization in some way.

Adam combines:

    Momentum
        +
    RMSProp
        ↓
       Adam

Adam keeps track of two things:

    First moment  → average of gradients
    Second moment → average of squared gradients

## Adam's Main Idea

Adam maintains two moving averages.

First:

    m = moving average of gradients

Second:

    v = moving average of squared gradients

Therefore:

    m → direction of gradients

    v → magnitude of gradients

Adam uses both to determine
how the weights should be updated.

## Adam Parameters

### g

Current gradient.

    g = current gradient calculated by backpropagation


### m

First moment.

    m = moving average of gradients

This is related to Momentum.


### v

Second moment.

    v = moving average of squared gradients

This is related to RMSProp.


### beta1

Controls how much previous gradient information
is remembered.

Common value:

    beta1 = 0.9


### beta2

Controls how much previous squared-gradient
information is remembered.

Common value:

    beta2 = 0.999


### epsilon

Small value that prevents division by zero.

Typical value:

    epsilon = 1e-7

## Adam Combines Momentum and RMSProp

Momentum:

    m = moving average of gradients

RMSProp:

    v = moving average of squared gradients

Adam:

    m + v

Therefore:

    Adam
      =
    Momentum-like behavior
      +
    RMSProp-like adaptive learning rate

## Simple Adam Example

Suppose:

    weight = 5
    gradient = 2

Initial:

    m = 0
    v = 0

Parameters:

    beta1 = 0.9
    beta2 = 0.999
    learning_rate = 0.1

First moment:

    m = 0.9 × 0 + 0.1 × 2

    m = 0.2

Second moment:

    v = 0.999 × 0 + 0.001 × 2²

    v = 0.004

Adam then applies bias correction and
uses the corrected values to calculate
the weight update.

The important point is:

    Adam considers both:

    direction → m

    magnitude → v

```python
import numpy as np

# Initial weight
w = 5.0

# Hyperparameters
learning_rate = 0.1
beta1 = 0.9
beta2 = 0.999
epsilon = 1e-8

# Initial moments
m = 0.0
v = 0.0

gradients = [2, 2, 2, 2, 2]

for t, gradient in enumerate(gradients, start=1):

    # First moment
    m = beta1 * m + (1 - beta1) * gradient

    # Second moment
    v = beta2 * v + (1 - beta2) * gradient**2

    # Bias correction
    m_hat = m / (1 - beta1**t)
    v_hat = v / (1 - beta2**t)

    # Adam update
    update = learning_rate * m_hat / (
        np.sqrt(v_hat) + epsilon
    )

    # Update weight
    w = w - update

    print(
        f"Step {t}: "
        f"m={m:.4f}, "
        f"v={v:.4f}, "
        f"update={update:.4f}, "
        f"weight={w:.4f}"
    )
```
