# Introduction to Unsupervised Learning

Unsupervised Learning is a type of Machine Learning where the model learns patterns and structures from data **without a target variable or labeled output**.

In supervised learning, we have:

$$
X \rightarrow y
$$

where:

- `X` = Features
- `y` = Target / Label

In unsupervised learning, we only have:

$$
X
$$

There is no predefined target variable.

The model tries to discover hidden patterns, groups, relationships, or structures within the data.

```markdown
# Main Types of Unsupervised Learning

The major areas of unsupervised learning are:

## 1. Clustering

Groups similar observations together.

Examples:

- K-Means
- Hierarchical Clustering
- DBSCAN
- Gaussian Mixture Models

---

## 2. Dimensionality Reduction

Reduces the number of features while trying to preserve important information.

Examples:

- PCA
- t-SNE
- UMAP

---

## 3. Association Rule Learning

Discovers relationships between items or variables.

Examples:

- Apriori
- FP-Growth

A common application is:

**Market Basket Analysis**

Example:

```text
Customers who buy:
Bread + Butter
        ↓
may also buy:
Milk


---

# 5. Clustering

### Markdown Cell

```markdown
# Clustering

Clustering is an unsupervised learning technique that divides observations into groups called **clusters**.

Observations within the same cluster should be relatively similar.

Observations belonging to different clusters should be relatively different.

### Example

Customer data:

```text
Customer
   ↓
Age
Income
Spending
   ↓
Clustering Algorithm
   ↓
Customer Groups


---

# 6. K-Means Clustering

### Markdown Cell

```markdown
# K-Means Clustering

K-Means is one of the most commonly used clustering algorithms.

It divides observations into **K clusters**.

The value of `K` represents the number of clusters we want to create.

### Example

If:

```text
K = 3


---

# 7. Hierarchical Clustering

### Markdown Cell

```markdown
# Hierarchical Clustering

Hierarchical Clustering creates a hierarchy of clusters.

It can be visualized using a **dendrogram**.

There are two main approaches:

### Agglomerative Clustering

Starts with each observation as its own cluster and gradually combines similar clusters.

```text
Individual Points
       ↓
Small Clusters
       ↓
Larger Clusters
       ↓
One Large Cluster


---

# 8. DBSCAN

### Markdown Cell

```markdown
# DBSCAN

DBSCAN stands for:

**Density-Based Spatial Clustering of Applications with Noise**

DBSCAN groups observations based on the density of data points.

It can also identify noise or outliers.

### Important Parameters

- `eps`
- `min_samples`

### Advantages

- Can find irregularly shaped clusters
- Can identify noise
- Does not require the number of clusters to be specified beforehand

### Disadvantage

Performance can be affected when clusters have very different densities.

# Dimensionality Reduction

Dimensionality reduction is the process of reducing the number of features in a dataset while trying to preserve important information.

For example:

```text
100 Features
     ↓
PCA
     ↓
10 Features


---

# 10. Principal Component Analysis

### Markdown Cell

```markdown
# Principal Component Analysis (PCA)

PCA is a dimensionality reduction technique.

It transforms the original features into a smaller number of new features called **Principal Components**.

The principal components are directions that capture as much variance in the data as possible.

### Example

Suppose we have:

```text
10 Features


---

# 11. Association Rule Learning

### Markdown Cell

```markdown
# Association Rule Learning

Association Rule Learning discovers relationships between items or events.

It is commonly used in:

**Market Basket Analysis**

Example:

```text
Bread → Butter


---

# 12. Anomaly Detection

### Markdown Cell

```markdown
# Anomaly Detection

Anomaly Detection identifies observations that behave differently from the majority of observations.

An anomaly can also be called:

- Outlier
- Abnormal observation
- Rare observation

### Example

Suppose a bank normally sees transactions between:

```text
$10 → $500


---

# 13. Important Unsupervised Learning Algorithms

### Markdown Cell

```markdown
# Important Unsupervised Learning Algorithms

## Clustering

### K-Means
Groups observations into K clusters.

### Hierarchical Clustering
Builds a hierarchy of clusters.

### DBSCAN
Uses density to discover clusters and noise.

### Gaussian Mixture Model
Uses probability distributions to model clusters.

---

## Dimensionality Reduction

### PCA
Reduces dimensions while preserving maximum variance.

### t-SNE
Useful for visualizing high-dimensional data.

### UMAP
A dimensionality reduction technique useful for visualization and structure preservation.

---

## Anomaly Detection

### Isolation Forest
Detects unusual observations by isolating them.

### One-Class SVM
Learns the boundary around normal observations.

### Local Outlier Factor
Detects observations with unusually low local density.

# Characteristics of Unsupervised Learning

### No Target Variable

The dataset generally contains only input features.

```python
X
