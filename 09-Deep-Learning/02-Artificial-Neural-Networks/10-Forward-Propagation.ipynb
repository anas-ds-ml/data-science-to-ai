# Forward Propagation

Forward Propagation is the process of passing input data through
the neural network from the input layer toward the output layer
to generate a prediction.

The data moves in one direction:

Input Layer
     ↓
Hidden Layer
     ↓
Hidden Layer
     ↓
Output Layer
     ↓
Prediction

## Why Do We Need Forward Propagation?

The neural network needs to transform input data into a prediction.

During forward propagation, every layer performs two main steps:

1. Calculate the weighted sum and add bias.
2. Apply an activation function.

The basic equations are:

Z = XW + b

A = f(Z)

Where:

X = Input
W = Weights
b = Bias
Z = Weighted Sum
f = Activation Function
A = Activation / Output

# Forward Propagation Flow

Input Data
    ↓
Input Layer
    ↓
Weighted Sum + Bias
    ↓
Activation Function
    ↓
Hidden Layer
    ↓
Weighted Sum + Bias
    ↓
Activation Function
    ↓
Output Layer
    ↓
Prediction

# Step 1: Input Data

Suppose our neural network receives two input features:

x₁ = 2
x₂ = 3

Input vector:

X = [2, 3]

These values are passed into the first layer.

# Step 2: Calculate Weighted Sum

Suppose the first neuron has:

w₁ = 0.5
w₂ = 0.8

Bias:

b = 1

The weighted sum is:

z = w₁x₁ + w₂x₂ + b

Substitute the values:

z = (0.5 × 2) + (0.8 × 3) + 1

z = 1 + 2.4 + 1

z = 4.4

# Step 3: Apply Activation Function

Suppose we use ReLU.

ReLU(z) = max(0, z)

Our weighted sum is:

z = 4.4

Therefore:

ReLU(4.4) = 4.4

The neuron output is:

a = 4.4

# Forward Propagation Through Multiple Layers

Suppose our ANN has:

Input Layer
     ↓
Hidden Layer 1
     ↓
Hidden Layer 2
     ↓
Output Layer

The calculations are:

Layer 1:

Z₁ = XW₁ + b₁

A₁ = ReLU(Z₁)

Layer 2:

Z₂ = A₁W₂ + b₂

A₂ = ReLU(Z₂)

Output Layer:

Z₃ = A₂W₃ + b₃

ŷ = Sigmoid(Z₃)

# Hidden Layer Weights

Suppose:

W₁ =

[ 0.5  0.2 ]
[ 0.8  0.4 ]

Bias:

b₁ = [1, 1]

The weighted sum is:

Z₁ = XW₁ + b₁

After calculating Z₁, we apply ReLU:

A₁ = ReLU(Z₁)

```python
import numpy as np

X = np.array([[2, 3]])

W1 = np.array([
    [0.5, 0.2],
    [0.8, 0.4]
])

b1 = np.array([1, 1])

Z1 = X @ W1 + b1

A1 = np.maximum(0, Z1)

print("Z1:", Z1)
print("A1:", A1)
```

# Hidden Layer Output

The hidden layer produces:

A₁ = ReLU(Z₁)

This output becomes the input to the next layer.

Therefore:

Input
  ↓
Hidden Layer 1
  ↓
A₁

The value A₁ is not the final prediction.

It is an intermediate representation learned by the network.

# Output Layer

Now suppose the output layer has one neuron.

Output weights:

W₂ = [0.6, 0.9]

Output bias:

b₂ = 0.5

The output layer calculates:

Z₂ = A₁W₂ + b₂

Then we apply the Sigmoid activation:

ŷ = Sigmoid(Z₂)

This gives the final prediction.

```python
W2 = np.array([[0.6],
               [0.9]])

b2 = np.array([0.5])

Z2 = A1 @ W2 + b2

y_pred = 1 / (1 + np.exp(-Z2))

print("Output Weighted Sum:", Z2)
print("Prediction:", y_pred)
```

# Forward Propagation During Prediction

When the model is used after training:

Input
  ↓
Forward Propagation
  ↓
Output
  ↓
Prediction

The model uses the learned weights and biases to generate
the prediction.

There is no weight update during normal prediction/inference.
