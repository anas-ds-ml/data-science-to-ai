# Tensor Operations

TensorFlow provides many operations for manipulating and
performing mathematical calculations on tensors.

Important tensor operations include:

1. Addition
2. Subtraction
3. Multiplication
4. Division
5. Matrix Multiplication
6. Exponentiation
7. Sum
8. Mean
9. Maximum
10. Minimum
11. Reshaping
12. Concatenation
13. Stacking
14. Transpose
15. Type Conversion

These operations are fundamental to neural network computations.

# Creating Tensors

Before performing operations, we need tensors.

TensorFlow provides:

tf.constant()
tf.zeros()
tf.ones()
tf.random.normal()
tf.random.uniform()

```python
import tensorflow as tf

a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

print("A:", a)
print("B:", b)
```

# Tensor Addition

Two tensors with compatible shapes can be added element by
element.

Example:

[1, 2, 3]
+
[4, 5, 6]

=

[5, 7, 9]

```python
result = tf.add(a, b)

print(result)
```

# Tensor Subtraction

Tensor subtraction is performed element by element.

Example:

[4, 5, 6]
-
[1, 2, 3]

=

[3, 3, 3]

```python
result = tf.subtract(b, a)

print(result)
```

# Element-wise Multiplication

Element-wise multiplication multiplies corresponding elements.

Example:

[1, 2, 3]
×
[4, 5, 6]

=

[4, 10, 18]

```python
result = tf.multiply(a, b)

print(result)
```

# Tensor Division

Division is also performed element by element.

```python
a = tf.constant([10.0, 20.0, 30.0])
b = tf.constant([2.0, 4.0, 5.0])

result = tf.divide(a, b)

print(result)
```

# Matrix Multiplication

Matrix multiplication is one of the most important operations
in neural networks.

TensorFlow provides:

tf.matmul()

For a neural network layer, we commonly have:

Z = XW + b

Where:

X = Input
W = Weight Matrix
b = Bias
Z = Weighted Sum

```python
X = tf.constant([
    [1.0, 2.0],
    [3.0, 4.0]
])

W = tf.constant([
    [0.5, 0.2],
    [0.1, 0.7]
])

result = tf.matmul(X, W)

print(result)
```

# Exponentiation

Exponentiation raises each tensor element to a power.

```python
x = tf.constant([1.0, 2.0, 3.0])

result = tf.pow(x, 2)

print(result)
```

# Square Root

TensorFlow provides:

tf.sqrt()

to calculate the square root of tensor elements.

```python
x = tf.constant([1.0, 4.0, 9.0, 16.0])

result = tf.sqrt(x)

print(result)
```

# Absolute Value

tf.abs() returns the absolute value of each element.

```python
x = tf.constant([-5.0, -2.0, 0.0, 3.0, 7.0])

result = tf.abs(x)

print(result)
```

# Sum

tf.reduce_sum() calculates the sum of tensor elements.

```python
x = tf.constant([1, 2, 3, 4, 5])

result = tf.reduce_sum(x)

print(result)
```

# Mean

tf.reduce_mean() calculates the average of tensor elements.

```python
x = tf.constant([10.0, 20.0, 30.0, 40.0])

result = tf.reduce_mean(x)

print(result)
```

# Maximum

tf.reduce_max() finds the largest value.

```python
x = tf.constant([10, 5, 30, 15, 20])

result = tf.reduce_max(x)

print(result)
```

# Minimum

tf.reduce_min() finds the smallest value.

```python
x = tf.constant([10, 5, 30, 15, 20])

result = tf.reduce_min(x)

print(result)
```

# Argmax

tf.argmax() returns the index of the largest value.

Example:

[10, 50, 20]

The largest value is:

50

Its index is:

1

```python
x = tf.constant([10, 50, 20])

index = tf.argmax(x)

print(index.numpy())
```

# Reshape

tf.reshape() changes the shape of a tensor without changing
the number of elements.

Example:

Shape:

(2, 3)

Can become:

(3, 2)

Both contain:

6 elements.

```python
x = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

print("Original:")
print(x)

reshaped = tf.reshape(x, (3, 2))

print("\nReshaped:")
print(reshaped)
```

# Transpose

Transpose swaps the dimensions of a tensor.

For a matrix:

Shape:

(2, 3)

After transpose:

Shape:

(3, 2)

```python
x = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

result = tf.transpose(x)

print(result)
```

# Concatenation

Concatenation joins tensors together along an existing axis.

Tensor A:

[1, 2, 3]

Tensor B:

[4, 5, 6]

Concatenation:

[1, 2, 3, 4, 5, 6]

```python
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

result = tf.concat([a, b], axis=0)

print(result)
```

# Stacking

tf.stack() combines tensors by creating a new dimension.

```python
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

result = tf.stack([a, b])

print(result)
print("Shape:", result.shape)
```
