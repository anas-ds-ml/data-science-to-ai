# Matrix Multiplication in Neural Networks

Neural networks use matrix multiplication to calculate the
weighted sum of inputs.

The basic equation of a neural network layer is:

Z = XW + b

Where:

X = input matrix
W = weight matrix
b = bias
Z = output before activation

After calculating Z, an activation function such as ReLU or
Sigmoid is applied.

A neural network performs this operation repeatedly across
different layers.

## 1. One Neuron

Suppose our neuron receives 3 inputs:

x₁ = 2
x₂ = 4
x₃ = 6

The neuron has 3 weights:

w₁ = 0.5
w₂ = 0.2
w₃ = 0.1

And a bias:

b = 1

The neuron calculates:

z = x₁w₁ + x₂w₂ + x₃w₃ + b

Therefore:

z = (2 × 0.5) + (4 × 0.2) + (6 × 0.1) + 1

z = 1 + 0.8 + 0.6 + 1

z = 3.4

This is a dot product plus a bias.

```python
import numpy as np

x = np.array([2, 4, 6])
w = np.array([0.5, 0.2, 0.1])
b = 1

z = np.dot(x, w) + b

print("Weighted sum:", z)
```

## 3. Matrix Multiplication Rule

For matrix multiplication:

A × B

the number of columns of A must equal the number of rows of B.

Example:

(1 × 3) × (3 × 2)

The inner dimensions are:

3 and 3

They match, so multiplication is possible.

The output shape is:

(1 × 2)

Therefore:

(1 × 3) × (3 × 2) = (1 × 2)

```python
X = np.array([
    [2, 4, 6]
])

W = np.array([
    [0.5, 0.2],
    [0.3, 0.4],
    [0.1, 0.6]
])

Z = X @ W

print("X shape:", X.shape)
print("W shape:", W.shape)
print("Z shape:", Z.shape)

print("\nZ:")
print(Z)
```

## 6. Adding Bias

Our matrix multiplication produced:

XW = [2.8  5.6]

Suppose our bias vector is:

b = [1.0  0.5]

Then:

Z = XW + b

Z = [2.8  5.6] + [1.0  0.5]

Z = [3.8  6.1]

So the complete operation is:

Z = XW + b

```python
X = np.array([
    [2, 4, 6]
])

W = np.array([
    [0.5, 0.2],
    [0.3, 0.4],
    [0.1, 0.6]
])

b = np.array([1.0, 0.5])

Z = X @ W + b

print("Z =", Z)
```

## 7. Applying Activation Function

The output of the linear operation is:

Z = [3.8  6.1]

Now we apply ReLU:

ReLU(z) = max(0, z)

Since both values are positive:

A = [3.8  6.1]

The complete layer is:

X
↓
XW
↓
XW + b
↓
ReLU
↓
A

```python
A = np.maximum(0, Z)

print("Before ReLU:", Z)
print("After ReLU :", A)
```
