# Recurrent Neural Networks (RNN)

Recurrent Neural Networks (RNNs) are a type of Deep Learning neural network designed to work with **sequential data**.

Unlike traditional neural networks, RNNs can use information from previous steps while processing the current step.

RNNs are especially useful when the **order of data matters**.

Examples of sequential data include:

- Text and sentences
- Speech and audio
- Time-series data
- Stock prices
- Weather measurements
- Sensor data
- User activity sequences
- DNA sequences

The main idea of an RNN is:

> **The current output depends on the current input and information remembered from previous time steps.**

## What is Sequential Data?

Sequential data is data where the **order of observations is important**.

For example, consider the sentence:

> "I love machine learning."

The words appear in a specific order:

```text
I → love → machine → learning

````python
# A simple sequence

sequence = ["I", "love", "machine", "learning"]

for time_step, word in enumerate(sequence, start=1):
    print(f"Time step {time_step}: {word}")
````

## Why Do We Need RNNs?

Traditional neural networks generally treat inputs independently.

For example, if we provide several words to a traditional model, it does not naturally maintain a memory of what happened at previous positions.

RNNs were designed to handle this type of problem.

An RNN processes a sequence one element at a time:

```text
Input 1 → RNN → Hidden State
                    ↓
Input 2 → RNN → Hidden State
                    ↓
Input 3 → RNN → Hidden State
                    ↓
Input 4 → RNN → Output

````python
import tensorflow as tf

# Create a Simple RNN layer
rnn = tf.keras.layers.SimpleRNN(64)

print(rnn)
````

## RNN Input Shape

RNNs normally work with three-dimensional input:

```text
(batch_size, time_steps, features)

````python
import tensorflow as tf

# batch_size = 2
# time_steps = 5
# features = 10

x = tf.random.normal((2, 5, 10))

print("Input shape:", x.shape)
````

````python
import tensorflow as tf

# Create Simple RNN
rnn = tf.keras.layers.SimpleRNN(64)

# Input:
# 2 sequences
# 5 time steps
# 10 features per time step
x = tf.random.normal((2, 5, 10))

# Pass input through RNN
output = rnn(x)

print("Input shape :", x.shape)
print("Output shape:", output.shape)
````

## RNN Memory

The most important concept in an RNN is the **hidden state**.

The hidden state carries information from previous time steps.

For example:

```text
"I → love → machine → learning"
