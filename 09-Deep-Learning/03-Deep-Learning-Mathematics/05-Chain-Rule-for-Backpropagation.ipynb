# Chain Rule for Backpropagation

The chain rule is one of the most important mathematical concepts behind neural network training.

A neural network is made of many connected operations.

For example:

$$
x \rightarrow z \rightarrow \hat{y} \rightarrow L
$$

Where:

- $x$ = input
- $z$ = neuron output before activation
- $\hat{y}$ = prediction
- $L$ = loss

The final loss depends on the weights indirectly.

The chain rule allows us to calculate:

$$
\frac{\partial L}{\partial w}
$$

This tells us how changing a weight affects the final loss.

This is exactly what backpropagation needs.

## What is the Chain Rule?

Suppose we have:

$$
y=f(u)
$$

and:

$$
u=g(x)
$$

Therefore:

$$
x \rightarrow u \rightarrow y
$$

To calculate how $y$ changes with respect to $x$:

$$
\boxed{
\frac{dy}{dx}
=
\frac{dy}{du}
\frac{du}{dx}
}
$$

Instead of calculating the complete derivative directly, we multiply the derivatives along the path.

This is the basic idea behind backpropagation.

## Simple Chain Rule Example

Suppose:

$$
u=x^2
$$

and:

$$
y=3u
$$

Therefore:

$$
x \rightarrow u \rightarrow y
$$

We want:

$$
\frac{dy}{dx}
$$

Using the chain rule:

$$
\frac{dy}{dx}
=
\frac{dy}{du}
\frac{du}{dx}
$$

First:

$$
\frac{dy}{du}=3
$$

And:

$$
\frac{du}{dx}=2x
$$

Therefore:

$$
\frac{dy}{dx}=3(2x)
$$

$$
\boxed{
\frac{dy}{dx}=6x
}
$$

```python
x = 2

u = x ** 2
y = 3 * u

derivative = 6 * x

print("x =", x)
print("u =", u)
print("y =", y)
print("dy/dx =", derivative)
```

## Chain Rule in a Neural Network

Consider a simple neuron:

$$
z=wx+b
$$

Suppose the neuron directly produces the prediction:

$$
\hat{y}=z
$$

And the loss is:

$$
L=(\hat{y}-y)^2
$$

The dependency is:

$$
w \rightarrow z \rightarrow \hat{y} \rightarrow L
$$

We want to calculate:

$$
\frac{\partial L}{\partial w}
$$

Using the chain rule:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat{y}}
\frac{\partial \hat{y}}{\partial z}
\frac{\partial z}{\partial w}
}
$$

## Step 1 — Derivative of Loss

Our loss is:

$$
L=(\hat{y}-y)^2
$$

Differentiate with respect to $\hat{y}$:

$$
\frac{\partial L}{\partial \hat{y}}
=
2(\hat{y}-y)
$$

This tells us how the loss changes when the prediction changes.

## Step 2 — Derivative of Prediction

We defined:

$$
\hat{y}=z
$$

Therefore:

$$
\frac{\partial \hat{y}}{\partial z}=1
$$

This is because changing $z$ directly changes $\hat{y}$ by the same amount.

## Step 3 — Derivative of Neuron

Our neuron is:

$$
z=wx+b
$$

We want:

$$
\frac{\partial z}{\partial w}
$$

When differentiating with respect to $w$, $x$ and $b$ are treated as constants.

Therefore:

$$
\boxed{
\frac{\partial z}{\partial w}=x
}
$$

## Step 4 — Apply the Chain Rule

We have:

$$
\frac{\partial L}{\partial \hat{y}}
=
2(\hat{y}-y)
$$

$$
\frac{\partial \hat{y}}{\partial z}=1
$$

$$
\frac{\partial z}{\partial w}=x
$$

Multiply them:

$$
\frac{\partial L}{\partial w}
=
2(\hat{y}-y)(1)(x)
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}
=
2(\hat{y}-y)x
}
$$

This is the gradient of the loss with respect to the weight.

```python
x = 2
w = 3
b = 1
y = 10

# Forward pass
z = w * x + b
y_pred = z

# Loss
loss = (y_pred - y) ** 2

# Chain rule
dL_dy_pred = 2 * (y_pred - y)
dy_pred_dz = 1
dz_dw = x

dL_dw = dL_dy_pred * dy_pred_dz * dz_dw

print("Prediction:", y_pred)
print("Loss:", loss)

print("\ndL/dy_pred:", dL_dy_pred)
print("dy_pred/dz:", dy_pred_dz)
print("dz/dw:", dz_dw)

print("\nGradient dL/dw:", dL_dw)
```

## Understanding the Gradient

We obtained:

$$
\frac{\partial L}{\partial w}=-12
$$

The negative sign tells us that increasing the weight would decrease the loss in this situation.

Gradient descent uses:

$$
w_{new}
=
w_{old}
-
\eta
\frac{\partial L}{\partial w}
$$

Suppose:

$$
\eta=0.1
$$

Then:

$$
w_{new}
=
3-(0.1)(-12)
$$

$$
w_{new}=4.2
$$

Therefore:

$$
3 \rightarrow 4.2
$$

The weight increases because its gradient was negative.

```python
learning_rate = 0.1

new_w = w - learning_rate * dL_dw

print("Old weight:", w)
print("Gradient:", dL_dw)
print("Learning rate:", learning_rate)
print("New weight:", new_w)
```

## Chain Rule with an Activation Function

Real neural networks normally contain activation functions.

For example:

$$
z=wx+b
$$

Then:

$$
a=ReLU(z)
$$

Then:

$$
L=(a-y)^2
$$

Now the dependency becomes:

$$
w
\rightarrow
z
\rightarrow
a
\rightarrow
L
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
\frac{\partial z}{\partial w}
}
$$

Notice that there is now one additional derivative:

$$
\frac{\partial a}{\partial z}
$$

This is the derivative of the activation function.

## ReLU Derivative

The ReLU function is:

$$
ReLU(z)=\max(0,z)
$$

Its derivative is:

$$
\frac{\partial ReLU(z)}{\partial z}
=
\begin{cases}
1 & z>0\\
0 & z<0
\end{cases}
$$

In simple terms:

- If $z>0$, gradient passes through.
- If $z<0$, gradient becomes 0.

This derivative becomes part of the chain rule during backpropagation.

```python
def relu(x):
    return max(0, x)

def relu_derivative(x):
    if x > 0:
        return 1
    else:
        return 0

values = [-2, -1, 0, 1, 2]

for value in values:
    print(
        "x =", value,
        "| ReLU =", relu(value),
        "| Derivative =", relu_derivative(value)
    )
```

# Chain Rule → Backpropagation

Backpropagation is essentially the repeated application of the chain rule.

Consider a neural network:

$$
x
\rightarrow
z_1
\rightarrow
a_1
\rightarrow
z_2
\rightarrow
a_2
\rightarrow
L
$$

To find the gradient of the loss with respect to an early weight $w_1$, we multiply all the local derivatives along the path:

$$
\frac{\partial L}{\partial w_1}
=
\frac{\partial L}{\partial a_2}
\frac{\partial a_2}{\partial z_2}
\frac{\partial z_2}{\partial a_1}
\frac{\partial a_1}{\partial z_1}
\frac{\partial z_1}{\partial w_1}
$$

This is why the error signal can travel backward through the network.

**Backpropagation = efficiently applying the chain rule from the loss backward through the network.**

## Backpropagation Flow

Forward:

Input
↓
Weighted Sum
↓
Activation
↓
Prediction
↓
Loss

Backward:

Loss
↓
Derivative of Loss
↓
Derivative of Activation
↓
Derivative of Weighted Sum
↓
Gradient of Weight
↓
Optimizer
↓
Updated Weight
