# Weighted Sum in Artificial Neural Networks

The weighted sum is one of the first mathematical operations
performed inside an artificial neuron.

A neuron receives input values and their corresponding weights.

Each input is multiplied by its weight, and all the resulting
values are added together.

A bias is then added to the result.

The basic equation is:

z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b

This value z is then passed through an activation function.

## Why is it Called "Weighted" Sum?

It is called a weighted sum because every input is multiplied
by a weight before the values are added.

For example:

x₁ × w₁
x₂ × w₂
x₃ × w₃

Then:

Weighted Sum = x₁w₁ + x₂w₂ + x₃w₃

The weights determine how much influence each input has on
the final result.

## Basic Formula

For two inputs:

z = w₁x₁ + w₂x₂ + b

For three inputs:

z = w₁x₁ + w₂x₂ + w₃x₃ + b

For n inputs:

z = Σ(wᵢxᵢ) + b

Where:

xᵢ = Input
wᵢ = Weight
b = Bias
z = Weighted Sum

```python
x1 = 2
x2 = 4
x3 = 3

w1 = 0.5
w2 = 0.2
w3 = 0.8

b = 1

z = (x1 * w1) + (x2 * w2) + (x3 * w3) + b

print("Weighted Sum:", z)
```
