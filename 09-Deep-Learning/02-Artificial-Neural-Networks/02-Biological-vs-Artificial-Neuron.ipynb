# Biological Neuron vs Artificial Neuron

A biological neuron is a basic functional unit of the human nervous
system.

An artificial neuron is a mathematical model inspired by the behavior
of biological neurons.

Artificial Neural Networks were inspired by the way biological neurons
receive, process, and transmit information.

## What is a Biological Neuron?

A biological neuron is a specialized cell that receives, processes,
and transmits information using electrical and chemical signals.

The main parts of a biological neuron are:

- Dendrites
- Cell Body (Soma)
- Nucleus
- Axon
- Axon Terminals
- Synapses

## Structure of a Biological Neuron

A simplified biological neuron can be represented as:

Dendrites
    ↓
Cell Body (Soma)
    ↓
Axon
    ↓
Axon Terminals
    ↓
Synapse
    ↓
Next Neuron

Each part has a different role in receiving and transmitting
information.

# What is an Artificial Neuron?

An artificial neuron is a mathematical computational unit
inspired by the basic idea of biological neurons.

It receives input values, assigns weights to those inputs,
adds a bias, and applies an activation function.

The mathematical representation is:

z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b

Then:

a = f(z)

Where:

- x = Input
- w = Weight
- b = Bias
- z = Weighted Sum
- f = Activation Function
- a = Output

## Structure of an Artificial Neuron

A simplified artificial neuron works like this:

Input
  ↓
Multiply by Weights
  ↓
Weighted Sum
  ↓
Add Bias
  ↓
Activation Function
  ↓
Output

Mathematically:

Inputs → Weighted Sum + Bias → Activation → Output

## Biological Neuron vs Artificial Neuron

| Biological Neuron | Artificial Neuron |
|---|---|
| Dendrites | Inputs |
| Synaptic Strength | Weights |
| Cell Body | Mathematical Processing |
| Electrical/Chemical Signals | Numerical Values |
| Axon | Output |
| Firing Behavior | Activation Function |
| Learning through biological mechanisms | Learning by updating weights and biases |

## Similarity Between Biological and Artificial Neurons

Both systems follow a simplified concept of:

Receive Information
        ↓
Process Information
        ↓
Produce Output

Biological:

Signals → Neuron → Signal

Artificial:

Inputs → Mathematical Computation → Output
