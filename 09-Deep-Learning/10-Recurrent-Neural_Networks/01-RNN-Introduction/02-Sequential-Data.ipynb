# Sequential Data

Sequential data is data where the **order of observations is important**.

In sequential data, each observation can be related to previous or future observations.

Examples include:

- Text
- Speech
- Time-series data
- Stock prices
- Weather measurements
- Sensor readings
- User activity
- Video frames
- DNA sequences

Recurrent Neural Networks (RNNs) are designed to process this type of data.

## What Is Sequential Data?

Sequential data consists of observations arranged in a particular order.

For example, consider the sentence:

```text
I → love → machine → learning

````python
sentence = ["I", "love", "machine", "learning"]

for time_step, word in enumerate(sentence, start=1):
    print(f"Time Step {time_step}: {word}")
````


## Sequential vs Non-Sequential Data

The main difference is whether the order of observations carries meaningful information.

### Sequential Data

```text
A → B → C → D

## Time Steps

A **time step** represents one position in a sequence.

For the sentence:

```text
I → love → machine → learning

````python
sequence = ["I", "love", "machine", "learning"]

for t, value in enumerate(sequence, start=1):
    print(f"x_{t} = {value}")
````

## Sequence Length

**Sequence length** is the number of time steps in one sequence.

For example:

```text
I → love → machine → learning

````python
sequence = ["I", "love", "machine", "learning"]

sequence_length = len(sequence)

print("Sequence:", sequence)
print("Sequence Length:", sequence_length)
````

## Batch Size

When training an RNN, we usually process multiple sequences together.

The number of sequences processed together is called the **batch size**.

For example:

```text
Batch Size = 4

````python
import numpy as np

batch_size = 32
time_steps = 10
features = 8

x = np.random.randn(batch_size, time_steps, features)

print("Input shape:", x.shape)
````

## Understanding the RNN Input Shape

RNN input generally has the following structure:

```text
(batch_size, time_steps, features)
