# Stochastic Gradient Descent (SGD)

Stochastic Gradient Descent is an optimization algorithm used to train machine learning and deep learning models.

Unlike Batch Gradient Descent, which uses the entire dataset to calculate one gradient, SGD uses **one training example at a time**.

For each training example:

**One Sample**
↓
**Forward Pass**
↓
**Calculate Loss**
↓
**Backpropagation**
↓
**Calculate Gradient**
↓
**Update Weights**

Then it moves to the next sample.

## Why is it called Stochastic?

The word **stochastic** means involving randomness.

Instead of calculating the gradient using the entire dataset, SGD typically processes individual training examples, usually in a shuffled order.

Suppose we have:

$$
1000
$$

training examples.

SGD approximately performs:

$$
1\ sample
\rightarrow
1\ gradient
\rightarrow
1\ update
$$

Then:

$$
1\ sample
\rightarrow
1\ gradient
\rightarrow
1\ update
$$

and continues until all samples have been processed.

## SGD Formula

For a single training example, the weight update is:

$$
\boxed{
W_{new}
=
W_{old}
-
\eta
\nabla_W L_i
}
$$

Where:

- $W$ = weights
- $\eta$ = learning rate
- $L_i$ = loss for one training example
- $\nabla_W L_i$ = gradient of that individual example

Compare this with Batch Gradient Descent:

$$
W_{new}
=
W_{old}
-
\eta
\frac{1}{m}
\sum_{i=1}^{m}
\nabla_W L_i
$$

The major difference is:

**Batch GD → gradient from all samples**

**SGD → gradient from one sample**

```python
weight = 5.0
learning_rate = 0.1

gradients = [2, 4, 6]

for gradient in gradients:
    weight = weight - learning_rate * gradient
    print("Updated weight:", weight)
```

## SGD from Scratch

Let's train a simple linear model:

$$
\hat{y}=wx+b
$$

We want the model to learn:

$$
y=2x
$$

We will use:

$$
L=(\hat{y}-y)^2
$$

For one sample:

$$
\hat{y}=wx+b
$$

The gradients are:

$$
\frac{\partial L}{\partial w}
=
2(\hat{y}-y)x
$$

and:

$$
\frac{\partial L}{\partial b}
=
2(\hat{y}-y)
$$

We update the parameters immediately after processing each sample.

```python
import numpy as np

# Training data
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)

# Initial parameters
w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 100

for epoch in range(epochs):

    # Shuffle the training data
    indices = np.random.permutation(len(X))

    for i in indices:

        # Select ONE sample
        x_i = X[i]
        y_i = y[i]

        # Forward pass
        y_pred = w * x_i + b

        # Error
        error = y_pred - y_i

        # Loss
        loss = error ** 2

        # Gradients
        dw = 2 * error * x_i
        db = 2 * error

        # Update immediately
        w = w - learning_rate * dw
        b = b - learning_rate * db

print("Final weight:", w)
print("Final bias:", b)
```
