# Introduction to TensorFlow

TensorFlow is an open-source machine learning and deep learning
framework developed by Google.

It is widely used for:

- Building neural networks
- Training deep learning models
- Computer vision
- Natural language processing
- Time series
- Model deployment

TensorFlow provides tools for creating, training, evaluating,
and deploying machine learning models.

# Why Do We Use TensorFlow?

Building a neural network manually requires implementing many
mathematical operations such as:

- Matrix multiplication
- Gradients
- Backpropagation
- Weight updates
- Optimization

TensorFlow provides these operations and automatically handles
many of the calculations required for training neural networks.

Therefore, we can focus more on designing and training models
rather than implementing every mathematical operation manually.

# TensorFlow and Deep Learning

# TensorFlow vs Keras

TensorFlow is the deep learning framework.

Keras is the high-level API used to build and train neural
networks easily.

Modern TensorFlow includes Keras as its primary high-level
deep learning API.

We commonly use:

import tensorflow as tf

and access Keras through:

tf.keras
